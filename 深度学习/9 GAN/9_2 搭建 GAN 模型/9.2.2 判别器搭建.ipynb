```python
import torch
import torch.nn as nn
```

```python
class Discriminator(nn.Module):
    def __init__(self, img_dim=784):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(img_dim, 1024),
            nn.LeakyReLU(0.2, True),
            nn.Dropout(0.3),

            nn.Linear(1024, 512),
            nn.LeakyReLU(0.2, True),
            nn.Dropout(0.3),

            nn.Linear(512, 256),
            nn.LeakyReLU(0.2, True),
            nn.Dropout(0.3),

            nn.Linear(256, 1),
            nn.Sigmoid()
        )

    def forward(self, img):
        return self.net(img)
```

```python
if __name__ == '__main__':
    D = Discriminator(img_dim=784)
    img = torch.randn(16, 784)
    prob = D(img)
    print(f"输入图片形状: {img.shape}")
    print(f"真假概率形状: {prob.shape}")
    print(f"概率值范围: {prob.min():.3f} ~ {prob.max():.3f}")
```
