```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
model = nn.Linear(784, 10)
```

```python
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

```python
x = torch.randn(32, 784)
```

```python
x
```

```python
labels = torch.randint(0, 10, (32,))

labels
```

```python
logits = model(x)
```

```python
logits
```

```python
loss = nn.CrossEntropyLoss()(logits, labels)
```

```python
loss
```

```python
loss.backward()
```

```python
optimizer.step()
```

```python
optimizer.zero_grad()
```

```python
print(f"SGD 更新完成，loss: {loss.item():.4f}")
```

```python
optimizer = optim.Adam(model.parameters(), lr=1e-3)

logits = model(x)
loss = nn.CrossEntropyLoss()(logits, labels)
loss
```

```python
loss.backward()
```

```python
optimizer.step()
```

```python
optimizer.zero_grad()
```

```python
print(f"Adam 更新完成，loss: {loss.item():.4f}")
```
