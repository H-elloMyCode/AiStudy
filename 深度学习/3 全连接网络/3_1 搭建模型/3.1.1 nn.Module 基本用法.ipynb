```python
import torch
import torch.nn as nn
```

```python
class SimpleNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(784, 128)
        self.layer2 = nn.Linear(128, 10)

    def forward(self, x):
        x = self.layer1(x)
        x = torch.relu(x)
        x = self.layer2(x)
        return x
```

```python
model = SimpleNet()
```

```python
x = torch.randn(4, 784)
x
```

```python
output = model(x)
```
