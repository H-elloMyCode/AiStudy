```python
import torch

t1 = torch.tensor([1, 2, 3])
print(t1)
```

```python
print(t1.dtype)
```

```python
t2 = torch.tensor([1.0, 2.0, 3.0], dtype=torch.float32)
print(t2)
```

```python
print(t2.dtype)
```

```python
zeros = torch.zeros(2, 3)
zeros
```

```python
print(zeros)
```

```python
ones = torch.ones(2, 3)
print(ones)
```

```python
like = torch.zeros_like(ones)
print(like)
```

```python
randn = torch.randn(2, 3)
print(randn)
```

```python
rand = torch.rand(2, 3)
print(rand)
```

```python
linspace = torch.linspace(0, 1, steps=5)

print(linspace)
```
