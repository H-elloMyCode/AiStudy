```python
from transformers import pipeline
```

```python
ner = pipeline("ner", aggregation_strategy="simple")
```

```python
text = "Elon Musk founded SpaceX in Hawthorn, California."
results = ner(text)
```

```python
results
```

```python
for entity in results:
    print(f"{entity['word']:20s} {entity['entity_group']:5s} (score: {entity['score']:.4f})")
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForTokenClassification
```

```python
model_name = "uer/roberta-base-finetuned-cluener2020-chinese"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name)
```

```python
model.eval()
```

```python
text = "张三在北京大学读书，毕业后去了腾讯工作。"
```

```python
inputs = tokenizer(text, return_tensors="pt")
```

```python
inputs
```

```python
with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits
```

```python
logits
```

```python
predictions = torch.argmax(logits, dim=-1)[0]
predictions
```

```python
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])
```

```python
tokens
```

```python
label_map = model.config.id2label
```

```python
label_map
```

```python
for token, pred_id in zip(tokens, predictions):
    label = label_map[pred_id.item()]
    if label != "O":
        print(f"{token:10s} → {label}")
```

```python
encoded = tokenizer(text, return_offsets_mapping=True)

offsets = encoded["offset_mapping"]

for offset, pred_id in zip(offsets, predictions):
    if offset[0] == 0 and offset[1] == 0:  # 跳过特殊 token
        continue
    label = label_map[pred_id.item()]
    if label != "O":
        entity_text = text[offset[0]:offset[1]]
        print(f"{entity_text:10s} → {label}")
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForTokenClassification

model_name = "ckiplab/bert-base-chinese-ner"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForTokenClassification.from_pretrained(model_name)
model.eval()

text = "张三在北京大学读书，毕业后去了腾讯工作。"
inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs)
    logits = outputs.logits  # (1, seq_len, num_labels)

predictions = torch.argmax(logits, dim=-1)[0]
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

label_map = model.config.id2label

for token, pred_id in zip(tokens, predictions):
    label = label_map[pred_id.item()]
    if label != "O":
        print(f"{token:10s} → {label}")
```
