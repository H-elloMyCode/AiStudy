```python
generator = pipeline("text-generation", model="gpt2")

result = generator("The future of AI is", max_length=50, num_return_sequences=1)
print(result[0]["generated_text"])
```

```python
from transformers import pipeline

ner = pipeline("ner", aggregation_strategy="simple")
result = ner("My name is John and I work at Google in New York.")
print(result)
```

```python
summarizer = pipeline(task="summarization")
text = """The Hugging Face transformers library provides thousands of pretrained models
to perform tasks on different modalities such as text, vision, and audio.
It has become the most popular library for NLP."""

result = summarizer(text, max_length=50, min_length=10)
print(result[0]["summary_text"])
```

```python
qa = pipeline("question-answering")
result = qa(
    question="What is the capital of France?",
    context="France is a country in Europe. Its capital is Paris, known for the Eiffel Tower."
)

print(result)
```

```python
from transformers import pipeline

summarizer = pipeline(task="summarization")
text = """The Hugging Face transformers library provides thousands of pretrained models
to perform tasks on different modalities such as text, vision, and audio.
It has become the most popular library for NLP."""

result = summarizer(text, max_length=50, min_length=10)
print(result[0]["summary_text"])
```

```python
from transformers import pipeline

summarizer = pipeline(task="summarization", device=0)

text = """The Hugging Face transformers library provides thousands of pretrained models
to perform tasks on different modalities such as text, vision, and audio.
It has become the most popular library for NLP."""

# 使用 max_new_tokens，控制生成摘要的token数量，避免警告
result = summarizer(
    text,
    max_new_tokens=25,
    min_new_tokens=10,
    do_sample=False
)
print(result[0]["summary_text"])
```

```python
qa = pipeline("question-answering")
result = qa(
    question="What is the capital of France?",
    context="France is a country in Europe. Its capital is Paris, known for the Eiffel Tower."
)

print(result)
```
