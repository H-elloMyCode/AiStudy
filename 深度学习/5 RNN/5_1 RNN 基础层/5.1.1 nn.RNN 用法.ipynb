```python
import torch
import torch.nn as nn
```

```python
rnn = nn.RNN(input_size=10, hidden_size=20, num_layers=1, batch_first=True)
```

```python
rnn
```

```python
x = torch.randn(3, 5, 10)
x
```

```python
output, h_n = rnn(x)
```

```python
output
```

```python
h_n
```

```python
print(f"输出形状: {output.shape}")
```

```python
print(f"最终隐藏状态: {h_n.shape}")
```
