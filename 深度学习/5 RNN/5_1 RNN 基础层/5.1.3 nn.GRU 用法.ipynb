```python
import torch
import torch.nn as nn
```

```python
gru = nn.GRU(input_size=10, hidden_size=20, num_layers=1, batch_first=True)
```

```python
gru
```

```python
x = torch.randn(3, 5, 10)
```

```python
x
```

```python
output, h_n = gru(x)
```

```python
output
```

```python
h_n
```

```python
print(f"输出形状: {output.shape}")
```

```python
print(f"隐藏状态: {h_n.shape}")
```

```python
gru_2layer = nn.GRU(input_size=10, hidden_size=20, num_layers=2, batch_first=True)
```

```python
output, h_n = gru_2layer(x)
```

```python
print(f"多层隐藏状态: {h_n.shape}")
```
