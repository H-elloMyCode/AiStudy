```python
import torch
from torch.nn.utils.rnn import pad_sequence
```

```python
word2idx = {"<pad>": 0, "<unk>": 1, "我": 2, "喜欢": 3, "深度学习": 4, "很": 5, "有趣": 6}
```

```python
word2idx
```

```python
encoded = [
    torch.tensor([2, 3, 4]),  # "我 喜欢 深度学习"
    torch.tensor([4, 5, 6]),  # "深度学习 很 有趣"
    torch.tensor([2, 3, 6, 4]),  # "我 喜欢 有趣 深度学习"
]
```

```python
encoded
```

```python
padded = pad_sequence(encoded, batch_first=True, padding_value=0)
```

```python
padded
```

```python
print(f"形状: {padded.shape}")
```

```python
lengths = torch.tensor([len(s) for s in encoded])
```

```python
lengths
```

```python
print(f"真实长度: {lengths}")
```

```python
from torch.nn.utils.rnn import pack_padded_sequence
```

```python
sorted_indices = lengths.argsort(descending=True)
sorted_indices
```

```python
padded
```

```python
sorted_padded = padded[sorted_indices]
```

```python
sorted_padded
```

```python
sorted_lengths = lengths[sorted_indices]
```

```python
sorted_lengths
```

```python
packed = pack_padded_sequence(sorted_padded, sorted_lengths.cpu(), batch_first=True)
packed
```

```python
print(f"packed 类型: {type(packed)}")
```
