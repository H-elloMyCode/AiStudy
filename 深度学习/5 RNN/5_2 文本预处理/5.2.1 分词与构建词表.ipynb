```python
from collections import Counter
```

```python
texts = ["我 喜欢 深度 学习", "深度 学习 很 有趣", "我 喜欢 有趣 的 东西"]
```

```python
texts
```

```python
tokenized = [text.split() for text in texts]
```

```python
tokenized
```

```python
counter = Counter(word for sent in tokenized for word in sent)
```

```python
counter
```

```python
print(counter.most_common(5))
```

```python
word2idx = {"<pad>": 0, "<unk>": 1}
word2idx
```

```python
for word, _ in counter.most_common():
    word2idx[word] = len(word2idx)
```

```python
word2idx
```

```python
idx2word = {v: k for k, v in word2idx.items()}
```

```python
idx2word
```

```python
print(f"词表大小: {len(word2idx)}")
```

```python
print(f"词表内容: {word2idx}")
```

```python
def encode(tokens, word2idx):
    return [word2idx.get(t, word2idx["<unk>"]) for t in tokens]
```

```python
sample = ["我", "喜欢", "深度", "学习"]
```

```python
print(f"编码结果: {encode(sample, word2idx)}")
```
