```python
import torch
import torch.nn as nn
```

```python
d_model = 64
nhead = 4
dim_ff = 128
num_layers = 3
seq_len = 10
batch_size = 2
```

```python
encoder_layer = nn.TransformerEncoderLayer(
    d_model=d_model,
    nhead=nhead,
    dim_feedforward=dim_ff,
    batch_first=True
)
```

```python
encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
```

```python
x = torch.randn(batch_size, seq_len, d_model)
```

```python
output = encoder(x)
```
