```python
import torch
from torch.utils.data import Dataset, DataLoader
from collections import Counter
```

```python
texts = [
    "这部电影真的很好看",
    "非常精彩的一部作品",
    "太无聊了浪费时间",
    "剧情很差不推荐",
    "演员演技很棒值得一看",
    "烂片一部浪费钱",
    "感人至深的好电影",
    "剧情拖沓毫无亮点",
]
```

```python
labels = [1, 1, 0, 0, 1, 0, 1, 0]
```

```python
def tokenize(text):
    return list(text)
```

```python
counter = Counter()
for text in texts:
    counter.update(tokenize(text))
```

```python
counter
```

```python
vocab = {word: idx + 2 for idx, (word, _) in enumerate(counter.most_common())}
```

```python
vocab
```

```python
vocab_size = len(vocab) + 2
```

```python
vocab_size
```

```python
def encode(text, max_len=15):
    tokens = [vocab.get(c, 1) for c in tokenize(text)]
    tokens = tokens[:max_len]
    tokens += [0] * (max_len - len(tokens))
    return tokens
```

```python
class TextDataset(Dataset):
    def __init__(self, texts, labels, max_len=15):
        self.texts = [encode(t, max_len) for t in texts]
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return torch.tensor(self.texts[idx]), torch.tensor(self.labels[idx])
```

```python
dataset = TextDataset(texts, labels)
```

```python
dataset
```

```python
dataloader = DataLoader(dataset, batch_size=4, shuffle=True)
```

```python
for batch_x, batch_y in dataloader:
    print(f"batch_x shape: {batch_x.shape}")
    print(f"batch_y shape: {batch_y.shape}")
    print(batch_x)
    print(batch_y)
    break
```
