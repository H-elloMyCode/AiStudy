```python
import torch
import torch.nn as nn
```

```python
x = torch.randn(2, 16, 8, 8)
```

```python
flatten = nn.Flatten()
out = flatten(x)
```

```python
print(f"展平前: {x.shape}")
```

```python
print(f"展平后: {out.shape}")
```

```python
flatten_partial = nn.Flatten(start_dim=2)
out_partial = flatten_partial(x)
print(f"只展平 H,W: {out_partial.shape}")
```

```python
model = nn.Sequential(
    nn.Conv2d(3, 16, 3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Flatten(),
    nn.Linear(16 * 16 * 16, 10)
)
```

```python
dummy = torch.randn(1, 3, 32, 32)
print(f"模型输出: {model(dummy).shape}")
```
