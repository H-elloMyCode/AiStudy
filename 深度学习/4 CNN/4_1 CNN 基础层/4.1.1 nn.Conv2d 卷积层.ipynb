```python
import torch
import torch.nn as nn
```

```python
x = torch.randn(1, 3, 32, 32)
x
```

```python
conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3)
```

```python
out1 = conv1(x)
```

```python
print(f"Conv2d(3, 16, 3) 输出形状: {out1.shape}")
```

```python
conv2 = nn.Conv2d(3, 16, kernel_size=3, padding=1)
out2 = conv2(x)
```

```python
print(f"Conv2d(3, 16, 3, padding=1) 输出形状: {out2.shape}")
```

```python
conv3 = nn.Conv2d(3, 32, kernel_size=3, stride=2, padding=1)
```

```python
out3 = conv3(x)
print(f"Conv2d(3, 32, 3, stride=2, padding=1) 输出形状: {out3.shape}")
```

```python
print(f"conv1 参数量: {sum(p.numel() for p in conv1.parameters())}")
```
