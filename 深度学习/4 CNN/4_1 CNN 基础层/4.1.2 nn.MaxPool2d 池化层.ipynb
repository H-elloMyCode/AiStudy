```python
import torch
import torch.nn as nn
```

```python
x = torch.randn(1, 16, 32, 32)
```

```python
print(x.shape)
```

```python
pool = nn.MaxPool2d(kernel_size=2)
out = pool(x)
print(f"MaxPool2d(2) 输出形状: {out.shape}")
```

```python
pool3 = nn.MaxPool2d(kernel_size=3, stride=3)
out3 = pool3(x)

print(f"MaxPool2d(3, stride=3) 输出形状: {out3.shape}")
```

```python
adaptive_pool = nn.AdaptiveAvgPool2d(output_size=(1, 1))
out_adaptive = adaptive_pool(x)

print(f"AdaptiveAvgPool2d(1,1) 输出形状: {out_adaptive.shape}")
```

```python
avg_pool = nn.AvgPool2d(kernel_size=2)
out_avg = avg_pool(x)
```

```python
print(f"MaxPool 和 AvgPool 输出形状相同: {out.shape} vs {out_avg.shape}")
```
