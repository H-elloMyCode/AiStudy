```python
import os
import shutil
import random

random.seed(42)
```

```python
src = os.path.expanduser(
    "C:\\Users\\LocalHost\\.cache\\kagglehub\\datasets\\alxmamaev\\flowers-recognition\\versions\\2\\flowers")
```

```python
dst = "data"
```

```python
classes = ["daisy", "rose", "sunflower"]
```

```python
train_num = 100
val_num = 20
```

```python
for cls in classes:
    src_dir = os.path.join(src, cls)
    all_files = os.listdir(src_dir)
    random.shuffle(all_files)

    os.makedirs(os.path.join(dst, "train", cls), exist_ok=True)
    os.makedirs(os.path.join(dst, "val", cls), exist_ok=True)

    for f in all_files[:train_num]:
        shutil.copy(os.path.join(src_dir, f), os.path.join(dst, "train", cls, f))

    for f in all_files[train_num:train_num + val_num]:
        shutil.copy(os.path.join(src_dir, f), os.path.join(dst, "val", cls, f))

print("数据整理完成")
```

```python
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
```

```python
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225]),
])
```

```python
val_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225]),
])
```

```python
train_dataset = datasets.ImageFolder("data/train", transform=train_transform)
val_dataset = datasets.ImageFolder("data/val", transform=val_transform)
```

```python
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=4)
```
