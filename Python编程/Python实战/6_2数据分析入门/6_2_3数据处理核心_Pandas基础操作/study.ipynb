##### pandas 中 loc 与 iloc 的核心区别

✅ loc 是「标签索引」，基于行 / 列的「标签名（label）」取值

✅ iloc 是「位置索引」，基于行 / 列的「数字位置（index position）」取值（从 0 开始）

```python
df = pd.DataFrame(
    data=np.arange(20).reshape(5, 4),
    index=[10, 20, 30, 40, 50],
    columns=['A', 'B', 'C', 'D']
)
df
```

```python
df.loc[30, 'B']
```

```python
df.iloc[2, 1]
```

```python
df.iloc[2, :]
```

```python
df.loc[20:40, 'B':'D']
```

```python
df.iloc[1:3, 1:3]
```

```python
df.loc[[10, 30, 50], ['A', 'C']]
```

```python
df.iloc[[0, 2, 4], [0, 2]]
```

```python
df.loc[(df['A'] > 5) & (df['B'] < 15), ['B', 'D']]
```

```python
test_df = pd.DataFrame(
    {"姓名": ["张三", "李四", "李四", "王五", None],
     "语文": [88, 95, None, 79, 92],
     "数学": [92, 87, 87, 90, 85],
     "英语": [78, 90, 90, None, 94]
     }
)
test_df
```

```python
test_df_drop_dup = test_df.drop_duplicates()
test_df_drop_dup
```

```python
test_df_drop_null = test_df_drop_dup.dropna()
test_df_drop_null
```

```python
test_df_fill_null = test_df_drop_dup.fillna(
    {
        "语文": test_df_drop_dup["语文"].mean(),
        "英语": 0,
        "姓名": "未知"
    }
)
test_df_fill_null
```
