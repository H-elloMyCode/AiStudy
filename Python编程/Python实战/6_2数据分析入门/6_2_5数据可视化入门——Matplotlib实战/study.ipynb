```python
import matplotlib.pyplot as plt
from matplotlib.backends.web_backend.nbagg_uat import colors
```

```python
# 配置中文字体（黑体，也可改为“微软雅黑”等系统支持的中文字体）
plt.rcParams["font.sans-serif"] = ["SimHei"]
# 解决负号“-”显示为方框的问题
plt.rcParams["axes.unicode_minus"] = False
```

```python
import pandas as pd
import numpy as np
```

```python
student_data = {
    "姓名": ["张三", "李四", "王五", "赵六", "孙七"],
    "语文": [88, 95, 79, 92, 85],
    "数学": [92, 87, 90, 85, 78],
    "英语": [78, 90, 86, 94, 82],
    "消费类别": ["餐饮", "购物", "交通", "娱乐", "其他"],
    "消费金额": [1500, 800, 300, 500, 200],
    "月度销量": [120, 150, 130, 180, 160]
}
student_df = pd.DataFrame(student_data)
student_df
```

```python
x_data = student_df["姓名"]
y_data = student_df["语文"]

plt.figure(figsize=(8, 5))

plt.bar(x_data, y_data, color="skyblue", width=0.6)

plt.xlabel("学生姓名", fontsize=10)
plt.ylabel("语文成绩", fontsize=10)
plt.title("5名学生语文成绩对比柱状图", fontsize=12, fontweight="bold")

for i in range(len(x_data)):
    plt.text(i, y_data[i] + 1, str(y_data[i]), ha="center", va="bottom", fontsize=9)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
x = np.arange(len(student_df["姓名"]))
width = 0.35

plt.bar(x - width / 2, student_df["语文"], width=width, label="语文", color="skyblue")
plt.bar(x + width / 2, student_df["数学"], width=width, label="数学", color="lightcoral")

plt.xticks(x, student_df["姓名"])

plt.xlabel("学生姓名", fontsize=10)
plt.ylabel("成绩", fontsize=10)
plt.title("5名学生语文&数学成绩对比柱状图", fontsize=12, fontweight="bold")

plt.legend()

for i in range(len(x_data)):
    plt.text(i - width / 2, student_df["语文"][i] + 1, str(student_df["语文"][i]), ha="center", va="bottom")
    plt.text(i + width / 2, student_df["数学"][i] + 1, str(student_df["数学"][i]), ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

```python
x_data = ["1月", "2月", "3月", "4月", "5月"]
y_data = student_df["月度销量"]
```

```python
plt.figure(figsize=(8, 5))

plt.plot(x_data, y_data, color="darkorange", linestyle="-", marker="o", markersize=8, label="月度销量")
plt.xlabel("月份", fontsize=10)
plt.ylabel("销量（件）", fontsize=10)
plt.title("5个月销量变化趋势折线图", fontsize=12, fontweight="bold")

plt.legend()

for i in range(len(x_data)):
    plt.text(i, y_data[i] + 2, str(y_data[i]), ha="center", va="bottom", fontsize=9)

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
labels = student_df["消费类别"]
sizes = student_df["消费金额"]

explode = (0.1, 0, 0, 0, 0)
```

```python
plt.figure(figsize=(7, 7))

wedges, texts, autotexts = plt.pie(
    sizes,
    explode=explode,
    labels= labels,
    autopct="%1.1f%%",
    textprops={"fontsize": 10},
    startangle=90
)

plt.axis("equal")

plt.title("个人消费类别金额占比饼图", fontsize=12, fontweight="bold")

for autotext in autotexts:
    autotext.set_color("white")
    autotext.set_fontweight("bold")

plt.tight_layout()
plt.show()
```
