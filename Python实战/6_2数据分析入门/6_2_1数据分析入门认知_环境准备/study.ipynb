```python
import numpy as np
```

```python
score_arr = np.array([85, 92, 78, 90, 88])
```

```python
avg_score = np.mean(score_arr)
max_score = np.max(score_arr)
```

```python
print("成绩数组: ", score_arr)
print(f'平均分: {avg_score}')
print(f'最高分: {max_score}')
```

```python
# 导入Matplotlib库（常用别名plt）
import matplotlib.pyplot as plt

# 解决中文显示问题（入门必备配置）
plt.rcParams["font.sans-serif"] = ["Kaiti"]  # 指定默认字体（黑体）
plt.rcParams["axes.unicode_minus"] = False  # 解决负号显示异常

# 准备数据（姓名和数学成绩）
names = ["张三", "李四", "王五"]
math_scores = [92, 87, 90]# 绘制柱状图
plt.bar(names, math_scores, color="skyblue")

# 设置图表标题和坐标轴标签
plt.title("学生数学成绩对比")
plt.xlabel("姓名")
plt.ylabel("数学成绩")

# 显示图表
plt.show()
```
