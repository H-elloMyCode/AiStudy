```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
student_data = {
    "姓名": ["张三", "李四", "王五", "赵六", "孙七", "周八", "吴九", "郑十"],
    "性别": ["男", "男", "女", "女", "男", "男", "女", "女"],
    "语文": [88, 95, 79, 92, 85, 76, 90, 83],
    "数学": [92, 87, 90, 85, 78, 94, 88, 91],
    "英语": [78, 90, 86, 94, 82, 75, 89, 92],
    "物理": [90, 85, 88, 92, 76, 95, 84, 89]
}
```

```python
student_df = pd.DataFrame(student_data)
student_df
```

```python
desc_stats = student_df.describe().round(2)
```

```python
desc_stats
```

```python
subject_cols = ["语文", "数学", "英语", "物理"]
```

```python
avg_scores = student_df[subject_cols].mean().round(2)
avg_scores
```

```python
max_scores = student_df[subject_cols].max()
max_scores
```

```python
min_scores = student_df[subject_cols].min()
min_scores
```

```python
std_scores = student_df[subject_cols].std().round(2)
std_scores
```

```python
median_scores = student_df[subject_cols].median().round(2)
median_scores
```

```python
student_df["总分"] = student_df[subject_cols].sum(axis=1)
student_df
```

```python
student_df["个人平均分"] = student_df[subject_cols].mean(axis=1).round(2)
student_df
```

```python
gender_count = student_df["性别"].value_counts()
gender_count
```

```python
gender_ratio = (student_df["性别"].value_counts(normalize=True) * 100).round(2)
gender_ratio
```

```python
gender_group_stas = student_df.groupby("性别")[subject_cols].mean().round(2)
gender_group_stas
```

```python
student_df
```

```python
numeric_cols = ["语文", "数学", "英语", "物理", "总分", "个人平均分"]
```

```python
corr_matrix = student_df[numeric_cols].corr().round(2)
corr_matrix
```

```python
math_corr = corr_matrix["数学"].sort_values(ascending=False)
math_corr
```

```python
plt.figure(figsize=(10, 8))
plt.imshow(corr_matrix, cmap='RdYlBu_r', aspect='auto', vmin=-1, vmax=1)
plt.colorbar(label="Pearson相关系数")

plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=45)
plt.yticks(range(len(corr_matrix.index)), corr_matrix.index)

for i in range(len(corr_matrix.index)):
    for j in range(len(corr_matrix.columns)):
        plt.text(j, i, corr_matrix.iloc[i, j], ha="center", va="center", fontsize=10)

plt.title("学生成绩变量相关系数热力图", fontsize=12)
plt.tight_layout()
plt.show()
```
