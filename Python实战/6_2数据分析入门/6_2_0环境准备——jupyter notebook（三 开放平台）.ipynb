```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 生成模拟数据
np.random.seed(0)  # 设置随机种子以确保结果可重复

# 假设有四个等级：A, B, C, D
grades = ['A', 'B', 'C', 'D']

# 随机生成100个学生的成绩
student_grades = np.random.choice(grades, 100, p=[0.2, 0.3, 0.4, 0.1])

# 将数据转换为DataFrame
grades_df = pd.DataFrame(student_grades, columns=['Grade'])

# 计算每个等级的学生数量
grade_counts = grades_df['Grade'].value_counts()

# 绘制饼图
plt.figure(figsize=(8, 8))
plt.pie(grade_counts, labels=grade_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Student Grades')
plt.axis('equal')  # 使饼图为圆形
plt.show()
```
