```python
import numpy as np
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
v = np.array([9, 10])
```

```python
mat_mul = np.linalg.multi_dot([A, B])
print(mat_mul)
```

```python
A_inv = np.linalg.inv(A)
print(A_inv)
```

```python
det_A = np.linalg.det(A)
print(det_A)
```

```python
x = np.linalg.solve(A, v)
print(x)
```

```python
eigenvals, eigenvecs = np.linalg.eig(A)
print(eigenvals)
```
