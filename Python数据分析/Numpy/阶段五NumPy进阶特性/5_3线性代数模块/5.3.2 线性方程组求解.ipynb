```python
import numpy as np
```

```python
A = np.array([[1, 2], [3, 4]])
b = np.array([9, 10])
```

```python
det_A = np.linalg.det(A)
print(det_A)
```

```python
x = np.linalg.solve(A, b)
print(x)
```

```python
verify = np.allclose(A @ x, b)
print(verify)
```

```python
A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]])
b = np.array([8, -11, -3])
```

```python
x = np.linalg.solve(A, b)
print(x)
```

```python
print(np.allclose(A @ x, b))
```

```python
A_error1 = np.array([[1, 2], [3, 4], [5, 6]])
b_error1 = np.array([7, 8, 9])
print(np.linalg.solve(A_error1, b_error1))
```

```python
A_error2 = np.array([[1, 2], [2, 4]])  # 行列式=0
b_error2 = np.array([3, 6])

print(np.linalg.det(A_error2))
```

```python
print(np.linalg.solve(A_error2, b_error2))
```

```python
def solve_linear(A, b):
    try:
        if A.shape[0] != A.shape[1]:
            raise ValueError('系数矩阵必须是方阵')
        x = np.linalg.solve(A, b)
        return x, np.allclose(A @ x, b)
    except Exception as e:
        return None, f"求解失败: {e}"

x, res = solve_linear(A_error2, b_error2)
print(res)
```
