```python
import numpy as np
```

```python
arr1 = np.ones((2, 3))
arr2 = np.ones(2)
```

```python
print(arr1)
```

```python
print(arr2)
```

```python
print(arr2.shape)
```

```python
print(arr1.shape)
```

```python
print(arr1 + arr2)
```

```python
arr2_fix = arr2.reshape(2, 1)
print(arr2_fix.shape)
```

```python
print(arr2_fix + arr1)
```

```python
arr_mat1 = np.ones((2, 3))
arr_mat2 = np.ones((1, 3))
print(arr_mat1 + arr_mat2)
```

```python
print(arr_mat1 @ arr_mat2)
```

```python
res_mat_mul = arr_mat1 @ arr_mat2.T
print(res_mat_mul)
```
