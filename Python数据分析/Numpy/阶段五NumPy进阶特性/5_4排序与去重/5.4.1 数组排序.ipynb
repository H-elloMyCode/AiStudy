```python
import numpy as np
```

```python
arr_1d = np.array([3, 1, 4, 1, 5, 9])
arr_2d = np.array([[5, 2, 9], [1, 7, 3]])
print(arr_1d)
```

```python
print(arr_2d)
```

```python
sorted_1d = np.sort(arr_1d)
print(sorted_1d)
```

```python
print(arr_1d)
```

```python
print(arr_2d)
```

```python
sorted_2d_row = np.sort(arr_2d, axis=1)
print(sorted_2d_row)
```

```python
sorted_2d_col = np.sort(arr_2d, axis=0)
print(sorted_2d_col)
```

```python
print(arr_2d)
```

```python
arr_1d_copy = arr_1d.copy()
print(arr_1d_copy)
```

```python
arr_1d_copy.sort()
```

```python
print(arr_1d_copy)
```

```python
print(arr_1d)
```

```python
index = np.argsort(arr_1d)
```

```python
print(index)
```

```python
sorted_by_index = arr_1d[index]
print(sorted_by_index)
```

```python
A = np.array([3, 1, 2])
B = np.array([10, 20, 30])
print(A)
```

```python
print(B)
```

```python
index_A = np.argsort(A)
print(index_A)
```

```python
sorted_A = A[index_A]
print(sorted_A)
```

```python
sorted_B = B[index_A]
```

```python
print(sorted_B)
```
