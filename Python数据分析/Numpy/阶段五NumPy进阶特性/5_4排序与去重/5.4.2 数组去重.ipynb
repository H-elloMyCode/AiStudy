```python
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([3, 4, 5, 6, 7])
print(arr1)
```

```python
print(arr2)
```

```python
diff = np.setdiff1d(arr1, arr2)
print(diff)
```

```python
intersect = np.intersect1d(arr1, arr2)
print(intersect)
```

```python
arr = np.array([3, 1, 4, 1, 5, 9, 5, 3])
print(arr)
```

```python
_, index = np.unique(arr, return_index=True)
unique_keep_order = arr[np.sort(index)]
print(unique_keep_order)
```
