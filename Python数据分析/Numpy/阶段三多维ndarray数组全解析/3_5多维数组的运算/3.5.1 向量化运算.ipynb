```python
import numpy as np
import time

arr = np.random.randint(0, 100, (1000, 1000))
print(arr)
```

```python
print(arr.shape)
```

```python
start = time.time()
res_vec = arr * 2
print("向量化耗时：", (time.time() - start) * 1000, "ms")
```

```python
start = time.time()
res_loop = np.zeros_like(arr)
for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        res_loop[i, j] = arr[i, j] * 2

print("循环耗时：", (time.time() - start) * 1000, "ms")
```
