```python
import numpy as np
```

```python
arr_a = np.array([[1, 2, 3], [4, 5, 6]])
arr_b = np.array([[7, 8], [9, 10], [11, 12]])
```

```python
print(arr_a)
```

```python
print(arr_b)
```

```python
res_dot = np.dot(arr_a, arr_b)
print("np.dot结果（2行2列）：\n", res_dot)
```

```python
res_at = arr_a @ arr_b
print("@运算符结果：\n", res_at)
```

```python
print(arr_a @ arr_a)
```

```python
import numpy as np

# 同形数组（2行2列）
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

# 1. 逐元素乘（向量化运算）
print("逐元素乘 * ：\n", arr1 * arr2)  # 输出：# [[ 5 12]#  [21 32]]

# 2. 矩阵乘法 @
print("矩阵乘法 @ ：\n", arr1 @ arr2)  # 输出：# [[19 22]#  [43 50]]
```

```python
print(arr1)
```

```python
print(arr2)
```

```python
arr_2d = np.array([[1, 2], [3, 4], [5, 6]])
print(arr_2d)
```

```python
arr_1d = np.array([7, 8])
print(arr_1d)
```

```python
print(arr_2d @ arr_1d)
```
