```python
import numpy as np
```

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
print(arr1)
```

```python
print(arr2)
```

```python
res_v = np.concatenate((arr1, arr2), axis=0)
print(res_v)
```

```python
res_h = np.concatenate((arr1, arr2), axis=1)
print(res_h)
```
