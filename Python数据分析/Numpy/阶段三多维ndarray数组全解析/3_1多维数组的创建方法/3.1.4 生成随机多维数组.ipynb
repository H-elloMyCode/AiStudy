```python
import numpy as np
```

```python
np.random.seed(666)

rand_2d = np.random.rand(2, 3)
print(rand_2d)
```

```python
np.random.seed(666)

randint_2d = np.random.randint(1, 10, (3, 2))
print(randint_2d)
```
