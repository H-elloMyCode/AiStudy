```python
import numpy as np
```

```python
arr_1d = np.array([1, 2, 3])
print(arr_1d)
```

```python
print(f"一维数组shape：{arr_1d.shape} → 仅axis=0")
```

```python
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d)
```

```python
print(f"二维数组shape：{arr_2d.shape} → axis0=行，axis1=列")
```

```python
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(f"三维数组shape：{arr_3d.shape} → axis0=层，axis1=行，axis2=列")
```

```python
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d)
```

```python
print(arr_2d.sum(axis=0))
```

```python
print(arr_2d.sum(axis=1))
```

```python
print(arr_2d.sum())
```
