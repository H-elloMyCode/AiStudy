```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

```python
print(arr.sum())
```

```python
print(arr.sum(axis=None))
```

```python
print(arr.mean(axis=None))
```

```python
print(arr.mean())
```

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

```python
res1 = arr.max(keepdims=False)
print(res1, type(res1))
```

```python
res2 = arr.max(keepdims=True)
print(res2, type(res2))
```

```python
print(res2.shape)
```

```python
arr_nan = np.array([1, np.nan, 3, np.nan, 5])
print(arr_nan)
```

```python
print(arr_nan.sum())
```

```python
print(arr_nan.sum(skipna=True))
```

```python
print(np.nansum(arr_nan))
```

```python
print(np.nanmean(arr_nan))
```

```python
arr = np.array([1, 2, 3, 4, 5])
arr_nan = np.array([1, np.nan, 3, 4])
print(arr)
print(arr_nan)
```

```python
print(arr.sum())
```

```python
print(arr_nan.mean())
```

```python
print(np.nanmean(arr_nan))
```

```python
print(arr.min(keepdims=True))
```
