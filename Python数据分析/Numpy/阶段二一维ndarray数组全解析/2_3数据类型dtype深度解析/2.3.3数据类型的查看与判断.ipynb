```python
arr_int = np.array([1, 2, 3])
arr_float = np.array([1.1, 2.2])

print(arr_int.dtype == int)
print(arr_float.dtype == float)
```

```python
print(np.issubdtype(arr_int.dtype, int))
```

```python
print(np.issubdtype(arr_float.dtype, float))
```
