```python
import numpy as np
```

```python
arr = np.array([[1.0, np.nan, 3.0], [4.0, np.nan, 5.0], [7.0, 8.0, 9.0]])
print(arr)
```

```python
arr_fill_0 = arr.copy()
print(arr_fill_0)
```

```python
arr_fill_0[np.isnan(arr_fill_0)] = 0
print(arr_fill_0)
```

```python
print(arr)
```

```python
arr_fill_mean = arr.copy()
print(arr_fill_mean)
```

```python
col_means = np.nanmean(arr_fill_mean, axis=0)
print(col_means)
```

```python
for i in range(arr_fill_mean.shape[1]):
    arr_fill_mean[np.isnan(arr_fill_mean[:, i]), i] = col_means[i]

print(arr_fill_mean)
```

```python
arr_fill_row = arr.copy()
print(arr_fill_row)
```

```python
row_means = np.nanmean(arr_fill_row, axis=1)
print(row_means)
```

```python
for i in range(arr_fill_row.shape[0]):
    arr_fill_row[i, np.isnan(arr_fill_row[i])] = row_means[i]

print(arr_fill_row)
```

```python
arr = np.array([[1.0, np.nan, 3.0], [4.0, 5.0, 6.0], [7.0, np.nan, 9.0]])
print(arr)
```

```python
arr_del_row = arr[~np.isnan(arr).any(axis=1)]
print(arr_del_row)
```

```python
print(arr[:, ~np.isnan(arr).any(axis=0)])
```

```python
arr_full_nan = np.array([[np.nan, np.nan], [1.0, 2.0]])
print(arr_full_nan)
```

```python
arr_del_full_row = arr_full_nan[~np.all(np.isnan(arr_full_nan), axis=1)]
print(arr_del_full_row)
```
