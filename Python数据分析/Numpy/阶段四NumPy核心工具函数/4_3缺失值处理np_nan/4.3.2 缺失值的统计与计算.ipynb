```python
arr = np.array([[1.0, np.nan, 3.0], [np.nan, 5.0, np.nan], [7.0, 8.0, 9.0]])
print(arr)
```

```python
print(np.isnan(arr).sum())
```

```python
print(np.isnan(arr).sum(axis=0))
```

```python
print(np.isnan(arr).sum() / arr.size)
```

```python
print(arr[np.isfinite(arr)].mean())
```
