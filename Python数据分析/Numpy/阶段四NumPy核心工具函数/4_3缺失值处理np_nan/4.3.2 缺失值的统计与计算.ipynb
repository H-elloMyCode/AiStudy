```python
arr = np.array([[1.0, np.nan, 3.0], [np.nan, 5.0, np.nan], [7.0, 8.0, 9.0]])
print(arr)
```

```python
nan_total = np.isnan(arr).sum()
print(nan_total)
```

```python
nan_col = np.isnan(arr).sum(axis=0)
print(nan_col)
```

```python
print(np.isnan(arr).sum(axis=1))
```

```python
nan_ratio = np.isnan(arr).sum() / arr.size
print(nan_ratio)
```

```python
valid_vals = arr[np.isfinite(arr)]
print(valid_vals)
```

```python
print(valid_vals.mean())
```
