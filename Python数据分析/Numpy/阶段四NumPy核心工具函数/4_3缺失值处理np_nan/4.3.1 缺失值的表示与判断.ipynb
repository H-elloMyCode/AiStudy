```python
import numpy as np
```

```python
arr_float = np.array([[1.0, 2.0, np.nan], [4.0, np.nan, 6.0]])
print(arr_float)
```

```python
arr_int = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_int)
```

```python
print(arr_int.dtype)
```

```python
arr_int[1, 1] = np.nan
print(arr_int)
```

```python
arr_int_float = arr_int.astype(float)
print(arr_int_float)
```

```python
arr_int_float[1, 1] = np.nan
print(arr_int_float)
```

```python
arr = np.array([[1.0, np.nan, 3.0], [np.inf, 5.0, -np.inf]])
print(arr)
```

```python
nan_mask = np.isnan(arr)
print(nan_mask)
```

```python
inf_mask = np.isinf(arr)
print(inf_mask)
```

```python
valid_mask = np.isfinite(arr)
print(valid_mask)
```

```python
print(arr)
```

```python
nan_count = np.isnan(arr).sum()
print(nan_count)
```
