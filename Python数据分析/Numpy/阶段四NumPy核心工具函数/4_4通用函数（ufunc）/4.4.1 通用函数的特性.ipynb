```python
import numpy as np
```

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([10, 20, 30])
```

```python
print(arr1)
```

```python
print(arr2)
```

```python
scalar = 2
```

```python
res_vec = np.multiply(arr1, scalar)
print(res_vec)
```

```python
res_broad = np.add(arr1, arr2)
print(res_broad)
```

```python
res_elem = np.sqrt(arr1)
print(res_elem)
```

```python
import time
```

```python
big_arr = np.random.rand(1000000)
print(big_arr)
```

```python
start = time.time()
ufunc_res = np.sin(big_arr)
print((time.time() - start) * 1000, "ms")
```

```python
print(ufunc_res)
```

```python
start = time.time()
loop_res = np.zeros_like(big_arr)
# print(loop_res)
for i in range(len(big_arr)):
    loop_res[i] = np.sin(big_arr[i])

print((time.time() - start) * 1000, "ms")
```

```python
print(loop_res)
```
