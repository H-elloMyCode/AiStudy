```python
import numpy as np
```

```python
def my_func(x):
    return x * 2 if x > 3 else x + 10
```

```python
print(my_func(3))
```

```python
print(my_func(4))
```

```python
print(my_func)
```

```python
ufunc_my = np.frompyfunc(my_func, 1, 1)
print(ufunc_my)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
```

```python
print(my_func(arr))
```

```python
res = ufunc_my(arr)
```

```python
print(res)
```

```python
res = res.astype(float)
print(res)
```

```python
def calc(x, y):
    return x * y if x + y > 10 else x - y
```

```python
print(calc(5, 6))
```

```python
print(calc(1, 2))
```

```python
vec_calc = np.vectorize(calc)
print(vec_calc)
```

```python
arr1 = np.array([[1, 2, 3], [4, 5, 6]])
arr2 = np.array([2, 4, 30])
```

```python
res = vec_calc(arr1, arr2)
print(res)
```
