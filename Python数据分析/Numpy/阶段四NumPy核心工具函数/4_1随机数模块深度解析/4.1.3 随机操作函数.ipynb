```python
import numpy as np
```

```python
np.random.seed(666)

arr1 = np.array([[1, 2], [3, 4], [5, 6]])
print(arr1)
```

```python
print(arr1)
```

```python
np.random.shuffle(arr1)
print(arr1)
```

```python
arr2 = np.array([10, 20, 30, 40])
print(arr2)
```

```python
perm_arr = np.random.permutation(arr2)
print(perm_arr)
```

```python
print(arr2)
```

```python
arr3 = np.array([1, 2, 3, 4, 5])
print(arr3)
```

```python
sample_no_replace = np.random.choice(arr3, size=3, replace=False)
print(sample_no_replace)
```

```python
sample_replace = np.random.choice(arr3, size=5, replace=True)
print(sample_replace)
```

```python
np.random.seed(666)

rand_sample = np.random.random_sample((2, 2))
print(rand_sample)
```

```python
weight_choice = np.random.choice([10, 20, 30], size=5, p=[0.1, 0.7, 0.2])
print(weight_choice)
```
