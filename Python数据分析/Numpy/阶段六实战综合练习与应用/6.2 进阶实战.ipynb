```python
import numpy as np
```

```python
ratings = np.array([
    [5, 3, np.nan, 1],
    [4, np.nan, 3, 2],
    [np.nan, 4, 2, 5],
    [2, 1, 5, 4]
])
```

```python
ratings_clean = ratings.copy()
print(ratings_clean)
```

```python
ratings_clean[np.isnan(ratings_clean)] = 0
print(ratings_clean)
```

```python
norm = np.linalg.norm(ratings_clean, axis=1, keepdims=True)
print(norm)
```

```python
normed_ratings = ratings_clean / norm
print(normed_ratings)
```

```python
similarity = normed_ratings @ normed_ratings.T
print(similarity)
```

```python
np.fill_diagonal(similarity, 0)
print(similarity)
```

```python
max_index = np.unravel_index(np.argmax(similarity), similarity.shape)
print(max_index)
```

```python
print(np.round(similarity, 2))
```

```python
print(f"用户{max_index[0] + 1} 和 用户{max_index[1] + 1}")
```

```python
print(f"相似度 {similarity[max_index]:.2f}")
```

```python
np.random.seed(42)
returns = np.random.randn(100, 5) * 0.01
print(returns)
```

```python
window = 5
print(returns.shape)
```

```python
roll_mean = np.array([np.mean(returns[i: i + window], axis=0) for i in range(len(returns) - window + 1)])
```

```python
print(roll_mean.shape)
```

```python
print(roll_mean)
```

```python
returns_copy = returns.copy()
print(returns_copy)
```

```python
std = np.std(returns_copy)
print(std)
```

```python
extremes = returns_copy[(returns_copy > 2 * std) | (returns_copy < -2 * std)]
print(extremes)
```

```python
unique_extremes = np.unique(extremes)
```

```python
print(unique_extremes)
```

```python
weekly_returns = returns.reshape(-1, 5, 5).sum(axis=1)
print(weekly_returns)
```

```python
print(np.round(roll_mean[:3], 4))
```

```python
print(np.round(unique_extremes, 4))
```

```python
print(np.round(weekly_returns[0], 4))
```
