#### 案例1

```python
sales = np.array([
    [100, 200, np.nan, 150],
    [300, -50, 250, 400],
    [np.nan, 180, 220, 190]
])
print(sales)
```

```python
sales_clean = sales.copy()
print(sales_clean)
```

```python
col_means = np.nanmean(sales_clean, axis=0)
print(col_means)
```

```python
for i in range(sales_clean.shape[1]):
    sales_clean[np.isnan(sales_clean[:, i]), i] = col_means[i]
print(sales_clean)
```

```python
sales_clean[sales_clean < 0] = np.nanmean(sales_clean, axis=0)[1]
print(sales_clean)
```

```python
print(sales_clean)
```

```python
total_sales = np.nansum(sales_clean)
print(total_sales)
```

```python
avg_sales = np.nanmean(sales_clean)
print(avg_sales)
```

```python
max_sales = np.nanmax(sales_clean)
print(max_sales)
```

#### 案例2

```python
scores = np.array([
    [85, 90, 78],
    [92, 88, 95],
    [76, 80, 82]
])
print(scores)
```

```python
mean = np.mean(scores, axis=0)
print(mean)
```

```python
std = np.std(scores, axis=0)
print(std)
```

```python
standardized = (scores - mean) / std
print(standardized)
```

#### 案例3

```python
consume = np.array([50, 30, 80, 30, 50, 100, 80, 50, 20, 100])
print(consume)
```

```python
unique_consume, counts = np.unique(consume, return_counts=True)
print(unique_consume)
```

```python
print(counts)
```

```python
sorted_index = np.argsort(unique_consume)[::-1]
sorted_consume = unique_consume[sorted_index]
print(sorted_consume)
```

```python
sorted_counts = counts[sorted_index]
print(sorted_counts)
```

```python
top3 = sorted_consume[:3]
print(top3)
```
