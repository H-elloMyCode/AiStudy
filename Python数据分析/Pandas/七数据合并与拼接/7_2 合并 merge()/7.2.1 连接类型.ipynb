```python
import pandas as pd
```

```python
path_main = r'order_main.csv'
path_product = r'product_info.csv'

df_order = pd.read_csv(path_main, encoding='utf-8')
df_product = pd.read_csv(path_product, encoding='utf-8')

df_order
```

```python
df_product
```

```python
merge_inner = pd.merge(df_order, df_product, on='商品ID', how='inner')

merge_inner
```

```python
merge_left = pd.merge(df_order, df_product, on='商品ID', how='left')

merge_left
```

```python
merge_right = pd.merge(df_order, df_product, on='商品ID', how='right')

merge_right
```

```python
merge_outer = pd.merge(df_order, df_product, on='商品ID', how='outer')

merge_outer
```
