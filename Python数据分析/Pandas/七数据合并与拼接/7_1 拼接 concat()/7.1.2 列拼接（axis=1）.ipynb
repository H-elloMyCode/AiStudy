```python
import pandas as pd
```

```python
file1_path = r'concat_col1.csv'
file2_path = r'concat_col2.csv'

df1 = pd.read_csv(file1_path, encoding='utf-8')
df2 = pd.read_csv(file2_path, encoding='utf-8')

df1
```

```python
df2
```

```python
df1 = df1.set_index('订单ID')
df2 = df2.set_index('订单ID')

df1
```

```python
df2
```

```python
df_concat = pd.concat([df1, df2], axis=1)

df_concat
```

```python
df_concat_inner = pd.concat([df1, df2], axis=1, join='inner')
df_concat_inner
```

```python
df2
```

```python
df_concat_lev = pd.concat([df1, df2],
                          axis=1,
                          keys=['基础信息', '补充信息'])

df_concat_lev
```

```python
df3 = pd.DataFrame(
    {
        '订单ID': [1001, 1002, 1003],
        '快递公司': ['顺丰', '圆通', '中通'],
        '发货时间': ['2024-01-01', '2024-01-02', '2024-01-03']
    }
).set_index('订单ID')

df3
```

```python
df_concat_multi = pd.concat([df1, df2, df3], axis=1)

df_concat_multi
```
