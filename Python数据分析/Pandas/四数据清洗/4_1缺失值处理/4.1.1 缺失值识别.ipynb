```python
import pandas as pd
import numpy as np
```

```python
file_path = r'missing_data.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
missing_flag = df.isnull()
missing_flag
```

```python
missing_count = df.isnull().sum()
missing_count
```

```python
missing_ratio = (df.isnull().sum() / len(df)) * 100
missing_ratio
```

```python
missing_info = pd.DataFrame(
    {
        '缺失数量': missing_count,
        '缺失占比（%）': missing_ratio.round(2)
    }
)

missing_info
```
