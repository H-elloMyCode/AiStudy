```python
import pandas as pd
```

```python
file_path = r'duplicate_drop.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(len(df))
```

```python
df_drop_default = df.drop_duplicates()

df_drop_default
```

```python
df_drop_last = df.drop_duplicates(keep='last')
df_drop_last
```

```python
df
```

```python
df_drop_id = df.drop_duplicates(subset=['订单ID'])

df_drop_id
```

```python
df_drop_multi = df.drop_duplicates(subset=['订单ID', '商品名称'], keep='last')

df_drop_multi
```

```python
df_drop_all = df.drop_duplicates(keep=False)

df_drop_all
```
