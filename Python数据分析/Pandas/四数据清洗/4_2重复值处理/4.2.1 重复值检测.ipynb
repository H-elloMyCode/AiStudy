```python
import pandas as pd
```

```python
file_path = r'duplicate_data.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(len(df))
```

```python
dup_flag = df.duplicated()
dup_flag
```

```python
dup_count = dup_flag.sum()

print(dup_count)
```

```python
print(len(df) - dup_count)
```

```python
dup_all_rows = df[df.duplicated(keep=False)]

dup_all_rows
```

```python
dup_only_rows = df[df.duplicated()]
dup_only_rows
```

```python
dup_by_id = df.duplicated(subset=['订单ID'])
dup_by_id
```

```python
dup_id_count = dup_by_id.sum()
print(dup_id_count)
```

```python
dup_by_multi = df.duplicated(subset=['订单ID', '商品名称'])
dup_by_multi
```

```python
dup_freq = df.groupby('订单ID').size().reset_index(name='重复次数')
dup_freq
```
