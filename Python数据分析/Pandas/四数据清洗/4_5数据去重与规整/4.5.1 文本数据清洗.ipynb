```python
import pandas as pd
```

```python
file_path = r'text_clean.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(df.loc[0].商品名称)
```

```python
print(df)
```

```python
df['商品名称'] = df['商品名称'].str.strip()
df['城市'] = df['城市'].str.strip()

df
```

```python
print(df.loc[0].商品名称)
```

```python
df['城市'] = df['城市'].str.replace(' ', '')
df
```

```python
df['备注'] = df['备注'].str.replace(r'[^\u4e00-\u9fa5a-zA-Z0-9]', '', regex=True).fillna('无备注')
df
```

```python
df['品牌'] = df['品牌'].str.strip().str.upper()

df
```

```python
df['版本'] = df['规格'].str.split('/', expand=True)[2]
df
```

```python
brand_map = {'HUAWEI': '华为', 'APPLE': '苹果', 'NIKE': '耐克', 'MI': '小米', 'LENOVO': '联想'}

df['品牌_中文'] = df['品牌'].map(brand_map)

df
```

```python
df_has_goods = df[df['备注'].str.contains('货', na=False)]

df_has_goods
```
