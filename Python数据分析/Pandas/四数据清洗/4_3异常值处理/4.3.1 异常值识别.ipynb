```python
file_path = r'outlier_data.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
price_outlier = df[(df['单价'] > 10000) | (df['单价'] < 0)]
qty_outlier = df[(df['数量'] > 100) | (df['数量'] < 1)]

price_outlier
```

```python
def detect_outlier_std(df_col):
    mean = df_col.mean()
    std = df_col.std()
    lower = mean - 3 * std
    upper = mean + 3 * std
    return df_col[(df_col < lower) | (df_col > upper)]

total_price_outlier = detect_outlier_std(df['总价'])
total_price_outlier
```

```python
q1 = df['数量'].quantile(0.25)
q3 = df['数量'].quantile(0.75)

iqr = q3 - q1
lower_q = q1 - 1.5 * iqr
upper_q = q3 + 1.5 * iqr

qty_outlier_iqr = df[(df['数量'] < lower_q) | (df['数量'] > upper_q)]
qty_outlier_iqr
```
