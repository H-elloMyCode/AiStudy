```python
import pandas as pd
import numpy as np
```

```python
file_path = r'outlier_handle.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
cond_delete = (df['数量'] > 100) | (df['单价'] > 10000)

df_delete = df[~cond_delete]
df_delete
```

```python
df
```

```python
df_fix = df.copy()

df_fix.loc[df_fix['数量'] > 100, '数量'] = 100
price_mean = df_fix[df_fix['单价'] <= 10000]['单价'].mean()

df_fix.loc[df_fix['单价'] > 10000, '单价'] = round(price_mean, 0)
df_fix['总价'] = df_fix['单价'] * df_fix['数量']

df_fix
```

```python
df
```

```python
df_replace = df.copy()

df_replace['数量异常标记'] = df_replace['数量'] > 100
df_replace['单价异常标记'] = df_replace['单价'] > 10000

df_replace
```

```python
q3_qty = df_replace['数量'].quantile(0.75)
iqr_qty = df_replace['数量'].quantile(0.75) - df_replace['数量'].quantile(0.25)

upper_qty = round(q3_qty + 1.5 * iqr_qty, 0)
df_replace.loc[df_replace['数量'] > upper_qty, '数量'] = upper_qty
df_replace
```

```python
df_replace['数量异常标记'] = df_replace['数量'] > 100
df_replace['单价异常标记'] = df_replace['单价'] > 10000

df_replace
```
