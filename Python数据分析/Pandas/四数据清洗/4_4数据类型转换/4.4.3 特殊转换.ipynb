```python
import pandas as pd
```

```python
file_path = r'dtype_special.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
price_map = {'低端': 0, '中端': 1, '中高端': 2}
df['价格区间编码'] = df['价格区间'].map(price_map)

df
```

```python
df
```

```python
df['用户等级编码'] = df['用户等级'].astype('category').cat.codes
df
```

```python
df
```

```python
tag_split = df['标签'].str.split('|', expand=True)

tag_split.columns = [f'标签{i + 1}' for i in range(tag_split.shape[1])]
tag_split
```

```python
df = pd.concat([df, tag_split], axis=1)
df
```

```python
df
```

```python
bins = [0, 50, 100, float('inf')]
labels = ['低销量', '中销量', '高销量']

df['销量等级'] = pd.cut(df['销量'], bins=bins, labels=labels)

df
```

```python
df
```

```python
df['是否高价值'] = (df['价格区间编码'] >= 1) & (df['销量'] > 50)
df['是否高价值_描述'] = df['是否高价值'].map({True: '是', False: '否'})

df
```
