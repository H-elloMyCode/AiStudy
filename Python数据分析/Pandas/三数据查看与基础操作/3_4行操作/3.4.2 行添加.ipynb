```python
import pandas as pd
```

```python
file_path = r'row_add.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(len(df))
```

```python
new_row1 = {'订单ID': 1004, '商品名称': '水杯', '单价': 59, '数量': 20, '城市': '深圳', '总价': 1180}
new_df = pd.DataFrame([new_row1])

df = pd.concat([df, new_df], ignore_index=True)

df
```

```python
new_row2 = [1005, '键盘', 899, 3, '北京', 2697]
df.loc[len(df)] = new_row2

df
```

```python
print(len(df))
```

```python
new_rows_df = pd.DataFrame(
    [
        [1006, '鼠标', 99, 8, '上海', 792],
        [1007, '背包', 199, 5, '广州', 995]
    ],
    columns=df.columns
)

df = pd.concat([df, new_rows_df], ignore_index=True)
df
```
