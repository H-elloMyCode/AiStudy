```python
import pandas as pd
```

```python
file_path = r'duplicate_data.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8'
)

df
```

```python
duplicate_flag = df.duplicated()
duplicate_flag
```

```python
duplicate_count = duplicate_flag.sum()
print(duplicate_count)
```

```python
duplicate_rows = df[duplicate_flag]
duplicate_rows
```

```python
print(len(df))
```

```python
print(f"重复行占比: {(duplicate_count / len(df) * 100):.2f}%")
```

```python
col_duplicate_flag = df.duplicated(subset=['商品名称', '销售额'], keep=False)
col_duplicate_flag
```

```python
col_duplicate_rows = df[col_duplicate_flag]
col_duplicate_rows
```

```python
print(col_duplicate_flag.sum())
```
