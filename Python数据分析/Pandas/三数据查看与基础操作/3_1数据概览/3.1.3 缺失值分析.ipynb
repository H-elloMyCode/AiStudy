```python
import pandas as pd
```

```python
file_path = r'missing_data.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['注册时间'],
    na_values=['']
)

df
```

```python
missing_count = df.isnull().sum()
missing_count
```

```python
missing_ratio = (df.isnull().sum() / len(df)) * 100
missing_ratio
```

```python
missing_analysis = pd.DataFrame(
    {'缺失值数量': missing_count,
     '缺失比例(%)': missing_ratio.round(2)}
)

missing_analysis
```
