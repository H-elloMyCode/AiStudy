```python
import pandas as pd
```

```python
file_path = r'data_modify_cond.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df['总价'] = df['总价'].astype('float64')
df['单价'] = df['单价'].astype('float64')

df
```

```python
df.loc[df['城市'] == '北京', '折扣'] = 0.8
df.loc[df['城市'] == '北京', '总价'] = df['单价'] * df['数量'] * df['折扣']

df
```

```python
cond = (df['城市'] == '上海') & (df['总价'] < 2000)

df.loc[cond, '单价'] = df.loc[cond, '单价'] * 1.1
df.loc[cond, '总价'] = df.loc[cond, '单价'] * df.loc[cond, '数量']

df
```

```python
df['客单等级'] = '低客单'
df
```

```python
df.loc[df['总价'] > 2000, '客单等级'] = '高客单'
df
```
