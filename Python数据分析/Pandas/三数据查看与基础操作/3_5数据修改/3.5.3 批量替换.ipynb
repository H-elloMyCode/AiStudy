```python
import pandas as pd
```

```python
file_path = r'data_replace.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
status_map = {'已完成': 'Completed', '待发货': 'Pending', '取消': 'Canceled'}
df['订单状态'] = df['订单状态'].replace(status_map)

df
```

```python
df['备注'] = df['备注'].replace(['缺货', '断货'], '库存不足')
df
```

```python
replace_dict = {
    '城市': {'北京': 'BJ', '上海': 'SH', '广州': 'GZ', '深圳': 'SZ'},
    '备注': {'无': '没有了', '库存不足': '没有库存了', '退款': '退款了'}
}

df.replace(replace_dict, inplace=True)
df
```

```python
df.replace('没有了', '无', inplace=True)
df
```
