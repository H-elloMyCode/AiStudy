```python
import pandas as pd
```

```python
file_path = 'column_add.csv'

df = pd.read_csv(file_path, encoding='utf-8')
df
```

```python
print(df.columns.tolist())
```

```python
df['订单状态'] = '已完成'
df
```

```python
df['支付方式'] = ['微信', '支付宝', '微信', '支付宝', '微信']
df
```

```python
df['总价'] = df['单价'] * df['数量']
df
```

```python
df['商品信息'] = df['商品名称'] + '（' + df['城市'] + '）'
df
```

```python
new_cols = {
    '折扣率': 0.9,
    '折扣总价': df['总价'] * 0.9,
    '利润': df['总价'] * 0.2
}

for col, value in new_cols.items():
    df[col] = value

df
```
