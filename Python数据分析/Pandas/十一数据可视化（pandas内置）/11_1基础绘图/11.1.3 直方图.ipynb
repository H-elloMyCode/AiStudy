```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.pyplot import xlabel
```

```python
# 设置中文显示（必加，避免乱码）
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
file_path = r'plot_hist.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8'
)

df
```

```python
df['销量'].plot(
    kind='hist',
    figsize=(10, 5),
    title='商品销量分布直方图',
    xlabel='销量区间',
    ylabel='频次(商品数量)',
    color='skyblue',
    edgecolor='black'
)

plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
df['销量'].plot(
    kind='hist',
    figsize=(10, 5),
    title='商品销量分布直方图',
    bins=[10, 15, 20, 25, 30, 35, 40],
    color='skyblue',
    edgecolor='black'
)

plt.xticks([10, 15, 20, 25, 30, 35, 40])
# plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
df
```

```python
df[['销量', '利润(元)']].plot(
    kind='hist',
    figsize=(10, 5),
    title='销量vs利润分布对比',
    bins=8,
    color=['orange', 'green'],
    edgecolor='black',
    alpha=0.7  # 透明度，避免重叠遮挡
)
plt.legend(loc='upper right')
plt.show()
```

```python
df
```

```python
df.hist(
    column='销量',
    by='品类',
    figsize=(12, 8),
    bins=5,
    color='lightgreen',
    edgecolor='black'
)

plt.suptitle('各品类销量分布', y=1.02)
plt.show()
```

```python
ax = df['销量'].plot(
    kind='hist',
    bins=6,
    figsize=(10, 5),
    color='purple',
    edgecolor='black'
)

mean_sales = df['销量'].mean()
median_sales = df['销量'].median()

ax.axvline(mean_sales, color='red', linestyle='--', label=f'均值：{mean_sales:.1f}')
ax.axvline(median_sales, color='blue', linestyle='--', label=f'中位数：{median_sales:.1f}')

ax.legend()
plt.title('销量分布（含均值/中位数）')
plt.show()
```

```python
df
```

```python
df['销量'].plot(
    kind='hist',
    figsize=(10, 5),
    title='销量分布（占比）',
    bins=6,
    density=True,
    color='gold',
    edgecolor='black'
)

plt.ylabel('占比')
plt.show()
```
