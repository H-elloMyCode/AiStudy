```python
from ctypes import kind

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.pyplot import xlabel, ylabel
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False    # 解决负号显示问题
```

```python
file_path = r'plot_line.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['日期'],
    index_col='日期'
)

df
```

```python
df['手机销量'].plot(
    kind='line',
    figsize=(10, 5),
    title='5月手机销量趋势',
    xlabel='日期',
    ylabel='销量',
    color='red'
)

plt.grid(True, alpha=0.3)
plt.savefig('手机销量折线图.png', dpi=100)
plt.show()
```

```python
df[['手机销量', '耳机销量']].plot(
    kind='line',
    figsize=(10, 5),
    title='5月手机/耳机销量对比',
    xlabel='日期',
    ylabel='销量',
    color=['red', 'blue'],
    marker='o'
)

plt.legend(loc='upper right')
plt.show()
```

```python
df
```

```python
fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.plot(df.index, df['手机销量'], color='red', marker='o', label='手机销量')
ax1.set_xlabel('日期')
ax1.set_ylabel('销量', color='red')
ax1.tick_params(axis='y', labelcolor='red')

ax2 = ax1.twinx()
ax2.plot(df.index, df['总销售额'], color='blue', marker='s', label='总销售额')
ax2.set_ylabel('销售额(元)', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1+lines2, labels1+labels2, loc='upper right')
plt.title('5月手机销量+销售额趋势')
plt.show()
```

```python
df['手机销量'].plot(
    kind='line',
    figsize=(10, 5),
    title='自定义折线样式',
    linestyle='--',
    linewidth=2,
    marker='^',
    markersize=8,
    alpha=0.5
)

plt.show()
```

```python
df_multi = df.reset_index()
df_multi
```

```python
df_multi['城市'] = '北京'
df_multi
```

```python
df_sh = df_multi.copy()
df_sh
```

```python
df_sh['城市'] = '上海'
df_sh['手机销量'] = [18,22,16,28,20,17,25]
df_sh['耳机销量'] = [13,16,10,18,14,12,17]

df_sh
```

```python
df_multi = pd.concat([df_multi, df_sh])

df_multi
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False    # 解决负号显示问题

df_multi.pivot(index='日期', columns='城市', values='手机销量').plot(
    kind='line',
    figsize=(10, 5),
    title='北京/上海手机销量对比'
)

plt.show()
```
