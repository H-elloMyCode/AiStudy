```python
import pandas as pd
```

```python
file_path = r'str_basic.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(df)
```

```python
df['商品名称_清洗'] = df['商品名称'].str.strip()

df
```

```python
print(df)
```

```python
df.to_csv('str_basic_1csv', encoding='utf-8', index=False)
```

```python
df
```

```python
df['品牌_清洗'] = df['品牌'].str.strip()

df
```

```python
df['规格描述_无空格'] = df['规格描述'].str.replace(' ', '')

df
```

```python
df['商品全称'] = df['商品名称_清洗'] + '-' + df['规格描述'].str.strip()

df
```

```python
df['备注_标注'] = '状态: ' + df['备注'].str.strip()

df
```

```python
df['品牌_en'] = ['Apple', 'Huawei', 'UNIQLO', 'Thermos', 'Rapoo', 'Logitech', 'Levi\'s', 'Grace']
df
```

```python
df['品牌_大写'] = df['品牌_en'].str.upper()
df['品牌_小写'] = df['品牌_en'].str.lower()

df
```

```python
df['备注长度'] = df['备注'].str.strip().str.len()

df
```

```python
df['规格前缀'] = df['规格描述'].str.strip().str[:5]
df
```

```python
df['备注_清洗'] = df['备注'].str.replace('  ', '、').str.strip()

df
```

```python
replace_map = {'热销': '畅销', '库存充足': '有货', '断码': '缺货'}

df['备注_替换'] = df['备注_清洗'].replace(replace_map)

df
```

```python
df['备注_final'] = df['备注_清洗'].replace('无', '暂无备注').fillna('暂无备注')
df
```

```python
df
```

```python
df_clean = df[['商品名称_清洗', '品牌_清洗', '规格描述_无空格', '商品全称', '备注_final', '城市']].rename(
    columns={'商品名称_清洗': '商品名称', '品牌_清洗': '品牌'})


df_clean
```
