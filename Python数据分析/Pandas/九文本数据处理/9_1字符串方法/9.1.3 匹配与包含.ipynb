```python
import pandas as pd
```

```python
file_path = r'str_match.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
df['含数码'] = df['商品名称'].str.contains('手机|耳机', na=False)

df
```

```python
digital_goods = df[df['含数码']]
digital_goods
```

```python
df['品牌含A'] = df['品牌'].str.contains('a', case=False, na=False)

df
```

```python
df['品牌以A开头'] = df['品牌'].str.match('^A', na=False)

df
```

```python
df['规格以钢结尾'] = df['规格'].str.endswith('钢', na=False)

df
```

```python
phone_pattern = r'^1[38]\d{9}$'

df['手机号格式正确'] = df['手机号'].astype(str).str.match(phone_pattern, na=False)

df
```

```python
df['备注含新品'] = df['备注'].str.contains('新品', na=False)

df
```

```python
df['一线城市'] = df['城市'].str.contains('北京|上海', na=False)

df
```

```python
df['无线+华为'] = df['规格'].str.contains('无线', na=False) & df['品牌'].str.contains('HW', na=False)

df
```

```python
df['备注_替换'] = df['备注'].str.replace('新品', 'NEW', regex=False)

df
```

```python
df['手机号_打码'] = df['手机号'].astype(str).str.replace(r'(\d{3})\d{4}(\d{4})', r'\1****\2', regex=True)
df
```
