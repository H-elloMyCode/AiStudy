```python
import pandas as pd
```

```python
file_path = r'str_split_join.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
df['规格列表'] = df['规格信息'].str.split('|')
df
```

```python
spec_split = df['规格信息'].str.split('|', expand=True)
spec_split
```

```python
spec_split.columns = ['规格1', '规格2', '规格3', '规格4']

df = pd.concat([df, spec_split], axis=1)

df
```

```python
df[['城市', '大区']] = df['城市_大区'].str.split('_', n=1, expand=True)

df
```

```python
df
```

```python
df['标签_拼接'] = df['标签'].str.split('|').str.join('、')
df
```

```python
df['商品_规格'] = df['商品名称'] + '(' + df['规格1'] + '-' + df['规格2'] + ')'

df
```

```python
df['主规格'] = df['规格列表'].str[0]

df
```

```python
df['最后标签'] = df['标签'].str.split('|').str[-1]

df
```

```python
def process_feedback(feedback):
    return '；'.join([x.strip() for x in feedback.split('|') if x])


df['反馈_清洗'] = df['用户反馈'].apply(process_feedback)

df
```

```python
df['是否无线'] = df['规格列表'].apply(lambda x: '无线' in x)

df
```

```python
df_null = df.copy()

df_null
```

```python
df_null.loc[2, '规格信息'] = None
df_null
```

```python
df_null['规格_split_safe'] = df_null['规格信息'].str.split('|', na=False)
df_null
```

```python
df_null['规格_split_safe'] = df_null['规格信息'].str.split('|').fillna('')
df_null
```
