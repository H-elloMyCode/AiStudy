```python
import pandas as pd
```

```python
file_path = r'str_slice_replace.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
df['编码后四位'] = df['商品编码'].str[4:]
df
```

```python
df['订单日期'] = df['订单号'].str[3:11]
df
```

```python
df['手机号后4位'] = df['手机号'].astype(str).str[-4:]
df
```

```python
df['编码奇数位'] = df['商品编码'].str[::2]
df
```

```python
df['备注_替换分隔符'] = df['备注'].str.replace('#', '、', regex=False)

df
```

```python
df['地址_替换'] = df['地址'].str.replace('市', '省', regex=False)
df
```

```python
df['备注_清洗'] = df['备注'].str.replace('#', '、').str.replace('热销', '畅销')
df
```

```python
df['手机号_脱敏'] = df['手机号'].astype(str).str.replace(r'(\d{3})\d{4}(\d{4})', r'\1****\2', regex=True)

df
```

```python
df['商品名称_数字'] = df['商品名称'].str.replace(r'\D', '', regex=True)

df
```

```python
replace_map = {'热销': '畅销', '断码': '缺货', '新品': '新品上市'}
df['备注_批量替换'] = df['备注_替换分隔符'].str.replace(replace_map, regex=False)

df
```

```python
df['省份'] = df['地址'].str[:2]
df
```

```python
df['省份_规范'] = (
    df['省份'].str.replace('京', '北京市').str.replace('沪', '上海市').str.replace('广', '广东省').str.replace('深',
                                                                                                               '广东省'))

df
```

```python
df
```

```python
df['备注_条件替换'] = df['备注_清洗'].apply(lambda x: x.replace('降价', '折扣') if '降价' in x else x)
df
```
