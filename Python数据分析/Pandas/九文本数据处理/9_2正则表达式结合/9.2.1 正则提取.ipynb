```python
import pandas as pd
```

```python
file_path = r'regex_extract.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
num_pattern = r'(\d+)'
df['商品数字'] = df['商品名称'].str.extract(num_pattern)

df
```

```python
order_pattern = r'-(\d+)'

df['订单号'] = df['订单信息'].str.extract(order_pattern)
df
```

```python
addr_pattern = r'(\d+号)'

df['门牌号'] = df['地址'].str.extract(addr_pattern)

df
```

```python
date_pattern = r'(\d{4}[-/.]?\d{2}[-/.]?\d{2})'

df['备注日期'] = df['备注'].str.extract(date_pattern)

df
```

```python
df['日期_规范'] = df['备注日期'].str.replace(r'[/.]', '-', regex=True)

df
```

```python
code_pattern = r'SP-(\d{4})-(\d+)'

df[['编码年份', '编号序号']] = df['编码'].str.extract(code_pattern)

df
```

```python
df['备注所有数字'] = df['备注'].str.findall(r'\d+')

df
```

```python
df
```

```python
import re

df['北京门牌号'] = df['地址'].apply(
    lambda x: re.search(r'(\d+号)', x).group(1) if '北京' in x else None
)
df
```
