```python
import pandas as pd
```

```python
file_path = r'regex_replace.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
phone_pattern = r'(\d{3})\d{4}(\d{4})'
df['手机号_脱敏'] = df['手机号'].astype(str).str.replace(phone_pattern, r'\1****\2', regex=True)

df
```

```python
addr_pattern = r'(\d+号).*'
df['地址_简化'] = df['地址'].str.replace(addr_pattern, r'\1', regex=True)

df
```

```python
date_pattern = r'(\d{4})[./](\d{2})[./](\d{2})'

df['备注_日期规范'] = df['备注'].str.replace(date_pattern, r'\1-\2-\3', regex=True)

df
```

```python
code_pattern = r'_'

df['编码_规范'] = df['编码'].str.replace(code_pattern, '-', regex=True)

df
```

```python
df['备注_清洗'] = df['备注_日期规范'].str.replace(r'&', '、', regex=True)

df
```

```python
status_pattern = r'(热销|库存充足|断码|新品|促销)'

df['备注_状态标注'] = df['备注_清洗'].str.replace(status_pattern, r'【状态: \1】', regex=True)

df
```

```python
order_pattern = r'^ORD-\d{8}-'

df['订单号_简化'] = df['订单号'].str.replace(order_pattern, '', regex=True)

df
```

```python
df_null = df.copy()

df_null
```

```python
df_null.loc[3, '手机号'] = 123456

df_null
```

```python
df_null['手机号_脱敏_safe'] = df_null['手机号'].astype(str).str.replace(r'(\d{3})\d{4}(\d{4})', r'\1****\2',
                                                                        regex=True,
                                                                        n=1  # 仅替换1次
                                                                        ).fillna('格式错误')

df_null
```

```python
import re


def desensitize_phone(phone):
    phone_str = str(phone).strip()

    # 严格匹配11位手机号
    if re.fullmatch(r'\d{11}', phone_str):
        return re.sub(r'(\d{3})\d{4}(\d{4})', r'\1****\2', phone_str)
    else:
        return '格式错误'


df_null['手机号_脱敏_safe2'] = df_null['手机号'].apply(desensitize_phone)
```
