```python
import pandas as pd
import numpy as np
```

```python
s1 = pd.Series([10, 20, 30, 40])
print(s1)
```

```python
s2 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(s2)
```

```python
arr = np.array([1.5, 2.5, 3.5])
print(arr)
```

```python
s3 = pd.Series(arr, index=['x', 'y', 'z'])
print(s3)
```

```python
dict_data = {'北京': 100, '上海': 200, '四川': 300}
print(dict_data)
```

```python
s4 = pd.Series(dict_data)
print(s4)
```

```python
s5 = pd.Series(dtype='float64')
print(s5)
```

```python
s6 = pd.Series()
print(s6)
```

```python
s6 = pd.Series([1, 2, 3], dtype='int32')
print(s6)
```
