```python
import pandas as pd
```

```python
file_path = r'rank_group.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
df['城市内销量排名'] = df.groupby('城市')['销量'].rank(ascending=False, method='min')
df
```

```python
df['品类内单价排名'] = df.groupby('品类')['单价'].rank(ascending=False, method='min')
df
```

```python
df['大区城市内总价排名'] = df.groupby(['大区', '城市'])['总价'].rank(ascending=False, method='dense')

df
```

```python
df[df['城市'] == '北京']
```

```python
df
```

```python
df['城市内销量排名_int'] = df['城市内销量排名'].astype(int)
df
```

```python
top2_by_city = df[df['城市内销量排名_int'] <= 2]
top2_by_city
```

```python
top2_by_city = top2_by_city.sort_values(['城市', '城市内销量排名_int'])

top2_by_city
```
