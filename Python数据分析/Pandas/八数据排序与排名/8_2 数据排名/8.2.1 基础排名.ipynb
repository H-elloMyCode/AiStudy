```python
import pandas as pd
```

```python
file_path = r'rank_basic.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
df['销量排名'] = df['销量'].rank(ascending=False)

df
```

```python
df['总价排名'] = df['总价'].rank(ascending=True)

df
```

```python
df_dup = df.copy()

df_dup
```

```python
df_dup['单价排名_平均'] = df_dup['单价'].rank(ascending=False, method='average')

df_dup
```

```python
df_dup['单价排名_最小'] = df_dup['单价'].rank(ascending=False, method='min')

df_dup
```

```python
df_dup['单价排名_最大'] = df_dup['单价'].rank(ascending=False, method='max')

df_dup
```

```python
df_dup['单价排名_连续'] = df_dup['单价'].rank(ascending=False, method='dense')

df_dup
```

```python
df
```

```python
df['销量排名_int'] = df['销量排名'].astype(int)
df
```

```python
top3_sales = df[df['销量排名_int'] <= 3]
top3_sales
```

```python
df['城市内总价排名'] = df.groupby('城市')['总价'].rank(ascending=False, method='min')

df
```
