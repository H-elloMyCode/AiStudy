```python
def filter_empty_groups(df, group_cols, min_rows=2, min_total=1000):
    group_obj = df.groupby(group_cols)
    filtered = group_obj.filter(lambda x: len(x) >= min_rows
                                          and x['总价'].sum() >= min_total
                                          and x.notna().all(axis=1).any())
    return filtered

final_filter = filter_empty_groups(df, ['大区', '品类'])
final_filter
```
