```python
import pandas as pd
```

```python
file_path = r'group_filter.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(len(df))
```

```python
filter1 = df.groupby('品类').filter(lambda x: x['总价'].sum() > 2000)
filter1
```

```python
print(len(filter1))
```

```python
filter2 = df.groupby('城市').filter(lambda x: x['单价'].mean() > 500)
filter2
```

```python
filter3 = df.groupby('大区').filter(lambda x: len(x) >= 3 and x['总价'].mean() > 2000)
filter3
```
