```python
import pandas as pd
```

```python
file_path = r'groupby_create.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(len(df))
```

```python
group_category = df.groupby('品类')
group_category
```

```python
for name, group in group_category:
    print(f"组名：{name} → 行数：{len(group)}")
```

```python
group_digital = group_category.get_group('数码')
group_digital
```

```python
df
```

```python
group_multi = df.groupby(['大区', '品类'])
group_multi
```

```python
print([name for name in group_multi.groups.keys()])
```

```python
group_north_digital = group_multi.get_group(('华北', '数码'))
group_north_digital
```
