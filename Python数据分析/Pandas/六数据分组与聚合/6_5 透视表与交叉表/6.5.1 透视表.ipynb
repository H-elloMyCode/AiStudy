```python
import pandas as pd
```

```python
file_path = r'pivot_table.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
pivot_basic = pd.pivot_table(
    df,
    index='品类',
    values='总价',
    aggfunc='sum'
)

pivot_basic
```

```python
pivot_multi = pd.pivot_table(
    df,
    index='大区',
    columns='品类',
    values='总价',
    aggfunc='sum',
    fill_value=0
)

pivot_multi
```

```python
pivot_multi_val = pd.pivot_table(
    df,
    index='城市',
    values=['数量', '总价'],
    aggfunc={'数量': 'sum', '总价': ['sum', 'mean']},
    fill_value=0
).round(2)

pivot_multi_val
```
