```python
import pandas as pd
```

```python
file_path = r'crosstab_data.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
ct_basic = pd.crosstab(df['品类'], df['是否包邮'])
ct_basic
```

```python
ct_margin = pd.crosstab(df['品类'], df['是否包邮'], margins=True, margins_name='总计')

ct_margin
```

```python
df
```

```python
ct_sum = pd.crosstab(df['大区'], df['客户等级'], values=df['销量'], aggfunc='sum')

ct_sum
```

```python
ct_mean = pd.crosstab(df['大区'], df['客户等级'], values=df['销量'], aggfunc='mean').round(2)

ct_mean
```

```python
df
```

```python
ct_row_multi = pd.crosstab([df['大区'], df['品类']], df['是否包邮'], margins=True)

ct_row_multi
```

```python
ct_col_multi = pd.crosstab(df['客户等级'], [df['品类'], df['是否包邮']])

ct_col_multi
```

```python
ct_row_pct = pd.crosstab(df['品类'], df['是否包邮'], normalize='index').round(2) * 100

ct_row_pct
```
