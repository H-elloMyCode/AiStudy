```python
import pandas as pd
```

```python
file_path = r'agg_rename.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
group_cat = df.groupby('品类')
group_cat
```

```python
agg_rename1 = group_cat['总价'].agg(品类总价合计='sum')
agg_rename1
```

```python
agg_rename2 = group_cat['总价'].sum().rename('品类总价求和')
agg_rename2
```

```python
agg_rename_multi = group_cat['总价'].agg(
    总价合计='sum',
    总价均值='mean',
    总价最大值='max',
    总价最小值='min'
).round(2)

agg_rename_multi
```

```python
agg_col_rename = group_cat.agg(
    单价最高值 = ('单价', 'max'),
    数量总计 = ('数量', 'sum'),
    总价中位数 = ('总价', 'median')
).round(2)

agg_col_rename
```

```python
group_multi = df.groupby(['大区', '品类'])
group_multi
```

```python
agg_multi_rename = group_multi['总价'].agg(大区品类总价合计 = 'sum').reset_index()
agg_multi_rename
```

```python
agg_temp = group_cat[['单价', '数量', '总价']].agg(['sum', 'mean'])
agg_temp
```

```python
agg_batch_flat = agg_temp.copy()
agg_batch_flat.columns = ['_'.join(col) for col in agg_batch_flat.columns]

# 再重命名
agg_batch_rename = agg_batch_flat.rename(columns={
    '单价_sum': '单价总价',
    '单价_mean': '单价均值',
    '数量_sum': '数量总和',
    '数量_mean': '数量均值',
    '总价_sum': '总价总和',
    '总价_mean': '总价均值'
}).round(2)

agg_batch_rename
```

```python
agg_batch_rename = agg_temp.rename(
    columns={
        ('单价', 'sum'): '单价总和',
        ('单价', 'mean'): '单价均值',
        ('数量', 'sum'): '数量总和',
        ('数量', 'mean'): '数量均值',
        ('总价', 'sum'): '总价总和',
        ('总价', 'mean'): '总价均值'}
).round(2)
agg_batch_rename
```
