```python
import pandas as pd
```

```python
file_path = r'agg_basic.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
group_cat = df.groupby('品类')
group_cat
```

```python
agg_sum = group_cat['总价'].sum()
agg_sum
```

```python
agg_mean = group_cat[['单价', '数量']].mean()
agg_mean.round(2)
```

```python
agg_multi_func = group_cat['总价'].agg(['sum', 'mean', 'max'])
agg_multi_func
```

```python
agg_rename = group_cat['总价'].agg(
    总价合计='sum',
    总价均值='mean',
    总价最大值='max'
).round(2)
agg_rename
```

```python
agg_col_func = group_cat.agg(
    单价最大值=('单价', 'max'),
    数量合计=('数量', 'sum'),
    总价均值=('总价', 'mean')
).round(2)

agg_col_func
```

```python
group_multi = df.groupby(['大区', '品类'])
group_multi
```

```python
agg_multi_group = group_multi['总价'].sum()
agg_multi_group
```
