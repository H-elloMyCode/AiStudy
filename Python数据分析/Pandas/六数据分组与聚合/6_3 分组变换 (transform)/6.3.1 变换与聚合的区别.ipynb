```python
import pandas as pd
```

```python
file_path = r'transform_vs_agg.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(len(df))
```

```python
group_cat = df.groupby('品类')
group_cat
```

```python
agg_sum = group_cat['总价'].agg('sum')
agg_sum
```

```python
print(len(agg_sum))
```

```python
df
```

```python
trans_sum = group_cat['总价'].transform('sum')
trans_sum
```

```python
print(len(trans_sum))
```

```python
df['品类总价合计（变换）'] = trans_sum
df
```

```python
agg_mean = group_cat['单价'].agg('mean').round(2)
agg_mean
```

```python
trans_mean = group_cat['单价'].transform('mean').round(2)
trans_mean
```

```python
df['品类单价均值（变换）'] = trans_mean
df
```

```python
df['品类内总价占比'] = (df['总价'] / df['品类总价合计（变换）']).round(4)
df
```

```python
df['单价标准化'] = group_cat['单价'].transform(lambda x: (x - x.mean()) / x.std()).round(4)
df
```

```python
df
```

```python
df[['品类单价最大值', '品类数量最大值']] = group_cat[['单价', '数量']].transform('max')
df
```
