```python
import pandas as pd
import numpy as np
```

```python
file_path = r'transform_scene.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(len(df))
```

```python
df['品类总价合计'] = df.groupby('品类')['总价'].transform('sum')
df
```

```python
df['城市数量均值'] = df.groupby('城市')['数量'].transform('mean').round(2)
df
```

```python
def z_score(x):
    return (x - x.mean()) / x.std()

df['单价标准化'] = df.groupby('品类')['单价'].transform(z_score).round(2)
df
```

```python
df
```

```python
df.loc[df['商品名称'].isin(['毛巾']), '单价'] = np.nan
df
```

```python
df['单价填充'] = df.groupby('品类')['单价'].transform(lambda x: x.fillna(x.mean()))
df
```

```python
df
```

```python
df['城市总价排名'] = df.groupby('城市')['总价'].transform(lambda x: x.rank(ascending=False, method='min')).astype(int)
df
```

```python
df
```

```python
df['组数量合计'] = df.groupby(['大区', '品类'])['数量'].transform('sum')
df
```

```python
df['数量占比'] = (df['数量'] / df['组数量合计']).round(2)
df
```

```python
agg_result = df.groupby('品类')['总价'].sum()
agg_result
```

```python
trans_result = df.groupby('品类')['总价'].transform('sum')
trans_result
```
