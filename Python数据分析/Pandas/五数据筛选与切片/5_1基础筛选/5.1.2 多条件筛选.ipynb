```python
import pandas as pd
```

```python
file_path = r'filter_multi.csv'

df = pd.read_csv(file_path, encoding='utf-8')
df
```

```python
df['下单时间'] = pd.to_datetime(df['下单时间'])
df
```

```python
print(len(df))
```

```python
cond_and = (df['单价'] > 200) & (df['城市'] == '北京') & (df['是否包邮'])
df_and = df[cond_and]

df_and
```

```python
cond_or = (df['城市'] == '上海') | (df['数量'] > 10) | (df['总价'] < 1000)
df_or = df[cond_or]

df_or
```

```python
cond_mix = df['是否包邮'] & ((df['单价'] > 500) | (df['下单时间'] >= '2024-01-04'))
df_mix = df[cond_mix]

df_mix
```

```python
cond_isin = df['城市'].isin(['北京', '上海']) & (df['总价'] > 1500)

df_isin = df[cond_isin]
df_isin
```
