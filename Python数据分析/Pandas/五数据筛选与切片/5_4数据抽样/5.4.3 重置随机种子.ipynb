```python
def stratified_sample(df, group_col, n, seed):
    np.random.seed(seed)
    return df.groupby(group_col).apply(lambda x: x.sample(n=n)).reset_index(drop=True)

sample_strat1 = stratified_sample(df, '城市', 1, 789)
sample_strat1
```

```python
sample_strat2 = stratified_sample(df, '城市', 1, 789)
sample_strat2
```
