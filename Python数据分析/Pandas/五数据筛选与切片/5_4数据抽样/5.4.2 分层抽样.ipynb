```python
import pandas as pd
```

```python
file_path = r'stratified_sample.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
print(df['城市'].value_counts())
```

```python
print(df['品类'].value_counts())
```

```python
df
```

```python
strat1 = df.groupby('城市').apply(lambda x: x.sample(n=1))
strat1
```

```python
strat2 = df.groupby('品类').apply(lambda x: x.sample(n=2))
strat2
```

```python
strat_frac = df.groupby('城市').apply(lambda x: x.sample(frac=0.3))
strat_frac
```

```python
df
```

```python
strat_multi = df.groupby(['城市', '品类']).apply(lambda x: x.sample(n=1) if len(x) >= 1 else x)
strat_multi
```

```python
strat_replace = df.groupby('品类').apply(lambda x: x.sample(n=3, replace=True))
strat_replace
```
