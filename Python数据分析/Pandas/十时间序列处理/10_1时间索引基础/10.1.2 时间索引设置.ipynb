```python
file_path = r'time_index.csv'
df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
df_time_idx['年'] = df_time_idx.index.year
df_time_idx['小时'] = df_time_idx.index.hour

df_time_idx
```

```python
df_split = df_reset.copy()

df_split
```

```python
df_split['日期'] = df_split['销售日期'].dt.date
df_split['时间'] = df_split['销售日期'].dt.time

df_split
```

```python
df_split['组合时间'] = pd.to_datetime(df_split['日期'].astype(str) + ' ' + df_split['时间'].astype(str))

df_split
```

```python
df_split_idx = df_split.set_index('组合时间')
df_split_idx
```
