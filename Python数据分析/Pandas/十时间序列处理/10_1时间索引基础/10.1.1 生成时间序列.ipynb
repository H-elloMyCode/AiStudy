```python
import pandas as pd
```

```python
file_path = r'time_series.csv'

df = pd.read_csv(file_path, encoding='utf-8')

df
```

```python
df['销售日期'] = pd.to_datetime(df['销售日期'])
df
```

```python
date_seq_basic = pd.date_range(start='2024-05-01', end='2024-05-08', freq='D')

date_seq_basic
```

```python
date_seq_days = pd.date_range(start='2024-05-01', periods=8, freq='D')

date_seq_days
```

```python
date_seq_hour = pd.date_range(start='2024-05-01', periods=6, freq='2h')

date_seq_hour
```

```python
date_seq_week = pd.date_range(start='2024-05-01', periods=4, freq='W-MON')

date_seq_week
```

```python
date_seq_month = pd.date_range(start='2024-01-01', periods=3, freq='ME')

date_seq_month
```

```python
df
```

```python
df_time_idx = df.set_index('销售日期')
```

```python
df_null = df_time_idx.reindex(date_seq_basic, fill_value="0")

df_null
```

```python
df_null = df_null.reset_index().rename(columns={'index': '销售日期'})
df_null
```
