```python
import numpy as np
import pandas as pd
```

```python
file_path = r'time_resample_down.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['销售时间'],
    index_col='销售时间'
)

df
```

```python
df_daily = df.resample('D')['销量'].sum()

df_daily
```

```python
df_weekly = df.resample('W-MON')['销量'].sum()

df_weekly
```

```python
df_monthly = df.resample('ME').agg({'销量': 'sum', '销售额': 'mean'})

df_monthly
```

```python
df_3days = df.resample('3D')['销量'].sum()

df_3days
```

```python
df_daily
```

```python
df_daily_fill = df.resample('D')['销量'].sum().dffillna(0)

df_daily_fill
```

```python
df_daily_fill = df_daily_fill.reset_index().rename(columns={'销售时间': '日期'})

df_daily_fill
```

```python
df_city_daily = df.groupby('城市').resample('D')['销量'].sum()

df_city_daily
```

```python
df_city_daily = df_city_daily.reset_index()

df_city_daily
```

```python
df_daily_stats = df.resample('D')['销量'].agg(['max', 'min', 'mean']).round(1)

df_daily_stats
```
