```python
import pandas as pd
```

```python
file_path = r'time_resample_up.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['销售日期'],
    index_col='销售日期'
)

df
```

```python
df_hourly = df.resample('h').asfreq()

df_hourly
```

```python
df_12h = df.resample('12h').asfreq()

df_12h
```

```python
df_hourly
```

```python
df_hourly_fill0 = df.resample('h').last().fillna(0)

df_hourly_fill0
```

```python
df_hourly_ffill = df.resample('h').ffill()

df_hourly_ffill
```

```python
df_hourly_bfill = df.resample('h').bfill()

df_hourly_bfill
```

```python
df_hourly_interp = df.select_dtypes(include=['number']).resample('h').interpolate(method='linear')

df_hourly_interp
```

```python
df_city_hourly = df.groupby('城市').resample('h').ffill()

df_city_hourly
```
