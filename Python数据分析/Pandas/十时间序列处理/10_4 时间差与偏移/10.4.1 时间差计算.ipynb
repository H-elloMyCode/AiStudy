```python
import pandas as pd
```

```python
file_path = r'time_delta_calc.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['下单时间', '支付时间', '发货时间', '签收时间']
)

df
```

```python
df['支付时长_分钟'] = (df['支付时间'] - df['下单时间']).dt.total_seconds() / 60

df
```

```python
df['发货时长_小时'] = (df['发货时间'] - df['支付时间']).dt.total_seconds() / 3600

df
```

```python
df['物流总时长'] = df['签收时间'] - df['发货时间']

df
```

```python
df
```

```python
df['物流时长_天'] = df['物流总时长'].dt.days
df['物流时长_小时'] = df['物流总时长'].dt.components.hours
df['物流时长_分钟'] = df['物流总时长'].dt.components.minutes

df
```

```python
df['订单全周期_天'] = (df['签收时间'] - df['下单时间']).dt.total_seconds() / (24 * 3600)

df
```

```python
df_long_logistics = df[df['物流时长_天'] >= 2]

df_long_logistics
```

```python
city_avg_logistics = df.groupby('城市')['订单全周期_天'].mean().round(2)

city_avg_logistics
```

```python
df
```

```python
max_pay_time = df['支付时长_分钟'].max()
min_logistics_time = df['物流总时长'].min()

max_pay_time
```

```python
min_logistics_time
```

```python
df['发货时长_延迟2h'] = df['发货时长_小时'] + 2

df
```

```python
df['预计签收时间'] = df['发货时间'] + pd.Timedelta(days=2)

df
```

```python
df['是否超时'] = df['签收时间'] > df['预计签收时间']

df
```
