```python
import pandas as pd
```

```python
file_path = r'data_shift.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['日期'],
    index_col='日期'
)

df
```

```python
df['昨日销量'] = df['日销量'].shift(1)
df
```

```python
df['明日销量'] = df['日销量'].shift(-1)

df
```

```python
df['2日前销量'] = df['日销量'].shift(2)

df
```

```python
df
```

```python
df['销量环比增长率(%)'] = ((df['日销量'] - df['昨日销量']) / df['昨日销量'] * 100).round(2)

df
```

```python
df['销售额环比差值'] = df['日销售额'] - df['日销售额'].shift(1)

df
```

```python
# df['上月同期销量'] = df['日销量'].shift(freq='ME')
#
# df
```

```python
df['3日销量均值'] = df['日销量'].rolling(window=3).mean().round(1)

df
```

```python
df['5日销量最大值'] = df['日销量'].rolling(window=5).max()

df
```

```python
df['3日销量总和'] = df['日销量'].rolling(window=3).sum()

df
```
