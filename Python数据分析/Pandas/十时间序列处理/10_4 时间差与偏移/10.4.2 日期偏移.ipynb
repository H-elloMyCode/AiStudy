```python
import pandas as pd
```

```python
file_path = r'time_offset.csv'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['下单日期']
)

df
```

```python
from pandas.tseries.offsets import Day, BusinessDay, MonthEnd, MonthBegin, Week, CustomBusinessDay
```

```python
df['下单+3天'] = df['下单日期'] + Day(3)

df
```

```python
df['下单+2工作日'] = df['下单日期'] + BusinessDay(2)

df
```

```python
df['下单-1周'] = df['下单日期'] - Week(1)

df
```

```python
df['当月月末'] = df['下单日期'] + MonthEnd(0)
df
```

```python
df['下月月初'] = df['下单日期'] + MonthBegin(1)

df
```

```python
df['本周周一'] = df['下单日期'] - pd.to_timedelta(df['下单日期'].dt.dayofweek, unit='d')

df
```

```python
holidays = [pd.Timestamp('2024-05-01')]

df['下单+2工作日_含节假日'] = df['下单日期'] + CustomBusinessDay(2, holidays=holidays)

df
```

```python
df['下单+2工作日_是否周末'] = df['下单+2工作日'].dt.dayofweek >= 5
df
```

```python
df_may_end = df[(df['下单日期'] >= (df['当月月末'] - Day(3))) & (df['下单日期'].dt.month == 5)]

df_may_end
```

```python
df['周一对齐'] = df['本周周一']
```

```python
week_amount = df.groupby('周一对齐')['付款金额'].sum()

week_amount
```

```python
df['5月31+1天'] = pd.Timestamp('2024-05-31') + Day(1)

df
```

```python
df['日期有效'] = df['下单+2工作日'].notna()

df
```

```python
df['工作日间隔'] = (df['下单+2工作日'] - df['下单日期']).dt.days

df
```
