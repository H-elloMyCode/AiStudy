```python
import pandas as pd
```

```python
file_path = r'time_dt_extract.csv.txt'

df = pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['销售时间']
)

df
```

```python
df['年'] = df['销售时间'].dt.year
df['月'] = df['销售时间'].dt.month
df['日'] = df['销售时间'].dt.day
df
```

```python
df['小时'] = df['销售时间'].dt.hour
df['分钟'] = df['销售时间'].dt.minute
df['秒'] = df['销售时间'].dt.second

df
```

```python
df['星期数'] = df['销售时间'].dt.dayofweek
df['星期名称'] = df['销售时间'].dt.day_name()
df['星期名称_中文'] = df['星期数'].map({0:'周一',1:'周二',2:'周三',3:'周四',4:'周五',5:'周六',6:'周日'})

df
```

```python
df['当月第几天'] = df['销售时间'].dt.day
df['当年第几天'] = df['销售时间'].dt.dayofyear

df
```

```python
df['是否周末'] = df['星期数'] >= 5
df
```

```python
df['仅日期'] = df['销售时间'].dt.date
df['仅时间'] = df['销售时间'].dt.time

df
```

```python
df['月份名称'] = df['销售时间'].dt.month_name()
df
```

```python
df
```

```python
df_time_idx = df.set_index('销售时间')
df_time_idx
```

```python
df_time_idx['索引小时'] = df_time_idx.index.hour
df_time_idx['索引星期'] = df_time_idx.index.dayofweek

df_time_idx
```
