```python
import pandas as pd
```

```python
file_path = r'time_timezone.csv'

df= pd.read_csv(
    file_path,
    encoding='utf-8',
    parse_dates=['下单时间(UTC)']
)

df
```

```python
df['UTC时间'] = df['下单时间(UTC)'].dt.tz_localize('UTC')

df
```

```python
df['北京时区'] = df['UTC时间'].dt.tz_convert('Asia/Shanghai')

df
```

```python
df['纽约时区'] = df['UTC时间'].dt.tz_convert('America/New_York')

df
```

```python
df['东京时区'] = df['UTC时间'].dt.tz_convert('Asia/Tokyo')

df
```

```python
df
```

```python
df['统一为北京时区'] = df.apply(lambda x: x['UTC时间'].dt.tz_convert('Asia/Shanghai'), axis=1)

df
```

```python
df['统一为北京时区'] = df['UTC时间'].dt.tz_convert('Asia/Shanghai')

df
```

```python
df['北京时区_小时'] = df['北京时区'].dt.hour

df['纽约时区_小时'] = df['纽约时区'].dt.hour

df
```

```python
local_time = pd.Series(['2024-05-01 08:00:00', '2024-05-01 09:00:00'])

local_time
```

```python
local_time_dt = pd.to_datetime(local_time)

local_time_dt
```

```python
beijing_local = local_time_dt.dt.tz_localize('Asia/Shanghai')

beijing_local
```

```python
beijing_to_utc = beijing_local.dt.tz_convert('UTC')

beijing_to_utc
```

```python
df['安全时区转换'] = df['下单时间(UTC)'].dt.tz_localize('UTC', ambiguous='NaT').dt.tz_convert('Asia/Shanghai')

df
```

```python
df['移除时区'] = df['北京时区'].dt.tz_localize(None)

df
```

```python
df['北京小时'] = df['统一为北京时区'].dt.hour

df
```
