```python
import pandas as pd

sales_data = {
    'product_name': ['手机', '耳机', '卫衣', '水杯'],
    'sales_num': [50, 80, 120, 200],
    'sale_date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    'city': ['北京', '上海', '广州', '深圳']
}

print(sales_data)
```

```python
sales_df = pd.DataFrame(sales_data)
sales_df
```

```python
from sqlalchemy import create_engine, INTEGER
```

```python
engine = create_engine('sqlite:///sales')
print(engine.name)
```

```python
print(engine.engine)
```

```python
sales_df.to_sql(
    name='sales',
    con=engine,
    if_exists='append',
    index=False,
    dtype={'sales_num': INTEGER}
)
```

```python
df_read_all = pd.read_sql('sales', con=engine)
df_read_all
```

```python
sql_query = "select product_name, sales_num, city from sales where sales_num > 60"
df_read_query = pd.read_sql(sql_query, con=engine)
df_read_query
```

```python
engine.dispose()
```
