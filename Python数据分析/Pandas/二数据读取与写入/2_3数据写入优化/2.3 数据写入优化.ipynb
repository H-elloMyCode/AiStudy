```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
test_data = {
    '订单ID': range(1001, 2001),
    '商品名称': np.random.choice(['手机', '耳机', '卫衣', '水杯'], 1000),
    '销售额': np.round(np.random.uniform(100, 5000, 1000), 2),
    '下单日期': pd.date_range('2024-01-01', periods=1000),
    '城市': np.random.choice(['北京', '上海', '广州', '深圳'], 1000)
}

print(test_data)
```

```python
df = pd.DataFrame(test_data)
```

```python
print(df)
```

```python
df
```

```python
df.to_csv(
    'sales_opt.csv',
    index=False,
    float_format='%.2f',
    encoding='utf-8',
    chunksize=500,
    compression='zip'
)
```

```python
with pd.ExcelWriter('sales_opt.xlsx') as writer:
    df.to_excel(
        writer,
        sheet_name='销售数据',
        index=False,
        merge_cells=False
    )
    df_summary = df.groupby('城市')['销售额'].sum().reset_index()
    df_summary.to_excel(writer, sheet_name='城市汇总', index=False)

print("✅ Excel优化写入完成！")
```

```python
from sqlalchemy import create_engine, INTEGER, FLOAT, DATE

engine = create_engine('sqlite:///pandas_sql')

df.to_sql(
    name='sales_opt',
    con=engine,
    if_exists='append',
    index=False,
    chunksize=100,
    dtype={
        '订单ID': INTEGER,
        '销售额': FLOAT,
        '下单日期': DATE
    }
)

engine.dispose()
```
