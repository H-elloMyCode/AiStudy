```python
import json

with open("dict_data.json", encoding="utf-8") as f:
    data = json.load(f)

# 遍历每个分类，用json_normalize拆解列表，再拼接
df = pd.concat(
    [pd.json_normalize(data[category]).assign(分类=category) for category in data.keys()],
    ignore_index=True
)[["分类", "商品ID", "价格"]]
```
