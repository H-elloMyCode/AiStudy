```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = np.linspace(-2, 2, 50)
y = np.linspace(-2, 2, 50)
```

```python
X, Y = np.meshgrid(x, y)
```

```python
Z = np.exp(-(X ** 2 + Y ** 2))
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection='3d')

ax.contour(X, Y, Z, 30, cmap='coolwarm')

ax.set_title("3D 等高线图", fontsize=14)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.tight_layout()

plt.show()
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection='3d')

ax.contour(X, Y, Z, 10, cmap='coolwarm')

ax.set_title("3D 等高线图", fontsize=14)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.tight_layout()

plt.show()
```
