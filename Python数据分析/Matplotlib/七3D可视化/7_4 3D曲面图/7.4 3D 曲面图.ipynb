```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = np.linspace(-2, 2, 50)
y = np.linspace(-2, 2, 50)
x
```

```python
y
```

```python
X, Y = np.meshgrid(x, y)
X
```

```python
Y
```

```python
Z = np.exp(-(X ** 2 + Y ** 2))
Z
```

```python
fig = plt.figure(figsize=(12, 6))

ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

ax.set_title("3D 曲面图", fontsize=14)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

ax.set_title("3D 曲面图", fontsize=14)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z 轴", fontsize=12, labelpad=15)

# plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

ax.set_title("3D 曲面图", fontsize=14)
ax.set_xlabel("X")
ax.set_ylabel("Y")
# ✅ 关键：调整视角，让 Z 轴标签不被遮挡
ax.view_init(elev=30, azim=-60)  # 俯仰角 30°，方位角 -60°

plt.subplots_adjust(left=0.1, right=0.9, bottom=0.1, top=0.9)  # 手动调整边距
plt.show()
# plt.show()
```
