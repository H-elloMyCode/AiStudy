```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
labels = ['力量', '速度', '技巧', '耐力', '智力']
values = [3.8, 4.2, 3.5, 4.0, 4.5]

N = len(labels)
```

```python
theta = np.linspace(0, 2 * np.pi, N, endpoint=False)
print(theta)

values = np.concatenate((values, [values[0]]))
print(values)

theta = np.concatenate((theta, [theta[0]]))
print(theta)

fig = plt.figure(figsize=(6, 6))

ax = fig.add_subplot(111, projection='polar')

ax.plot(theta, values, 'o-', linewidth=2, color='#FF6B6B')
ax.fill(theta, values, alpha=0.3, color='#FF6B6B')

ax.set_xticks(theta[:-1])
ax.set_xticklabels(labels)
ax.set_ylim(0, 5)

plt.title("雷达图（极坐标折线图）", fontsize=14)
plt.tight_layout()

plt.show()
```
