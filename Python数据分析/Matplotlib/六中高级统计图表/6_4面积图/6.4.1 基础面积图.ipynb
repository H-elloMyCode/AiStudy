```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5, 6, 7]
y = [2, 3, 4, 5, 4, 3, 2]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.fill_between(x, y, color='skyblue', alpha=0.5)


ax.set_title("基础面积图", fontsize=14)
ax.set_xlabel("X轴")
ax.set_ylabel("Y轴")
ax.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.fill_between(x, y, color='skyblue', alpha=0.5)
ax.plot(x, y, color='blue', linewidth=2)

ax.set_title("基础面积图", fontsize=14)
ax.set_xlabel("X轴")
ax.set_ylabel("Y轴")
ax.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```
