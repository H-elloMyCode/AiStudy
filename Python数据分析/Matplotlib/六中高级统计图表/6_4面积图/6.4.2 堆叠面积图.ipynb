```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5, 6]
y1 = [1, 2, 1, 2, 1, 2]
y2 = [2, 3, 4, 3, 2, 3]
y3 = [1, 2, 3, 2, 3, 1]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.stackplot(
    x,
    y1,
    y2,
    y3,
    labels=["A", "B", "C"],
    alpha=0.7
)

ax.set_title("堆叠面积图", fontsize=14)
ax.legend(loc="upper left")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
