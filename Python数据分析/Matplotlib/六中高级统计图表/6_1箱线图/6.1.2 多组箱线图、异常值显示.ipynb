```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
np.random.seed(20)

data1 = np.random.normal(50, 10, 200)
data2 = np.random.normal(60, 15, 200)
data3 = np.random.normal(70, 10, 200)
```

```python
all_data = [data1, data2, data3]
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(all_data, tick_labels=['组别A', '组别B', '组别C'])

ax.set_title("多组箱线图（含异常值）", fontsize=14)
ax.set_ylabel("数值分布")
ax.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
import numpy as np

# 固定种子（保证和你数据一样）
np.random.seed(20)

# 你的三组数据
data1 = np.random.normal(50, 10, 200)
data2 = np.random.normal(60, 15, 200)
data3 = np.random.normal(70, 10, 200)

# 定义一个函数，自动算箱线图所有指标
def boxplot_stats(data, name):
    Q1 = np.percentile(data, 25)
    Q2 = np.median(data)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = data[(data < lower) | (data > upper)]

    print(f"====== {name} 箱线图指标 ======")
    print(f"Q1 (25%):    {Q1:.2f}")
    print(f"中位数 (Q2): {Q2:.2f}")
    print(f"Q3 (75%):    {Q3:.2f}")
    print(f"IQR:         {IQR:.2f}")
    print(f"下界:        {lower:.2f}")
    print(f"上界:        {upper:.2f}")
    print(f"离群点数量:  {len(outliers)}")
    if len(outliers) > 0:
        print(f"离群点:      {np.round(outliers, 2)}")
    print("-" * 40)

# 计算三组数据
boxplot_stats(data1, "data1 (50,10)")
boxplot_stats(data2, "data2 (60,15)")
boxplot_stats(data3, "data3 (70,10)")
```
