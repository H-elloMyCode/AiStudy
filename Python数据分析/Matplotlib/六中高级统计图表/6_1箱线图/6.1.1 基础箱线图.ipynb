```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 生成和你一样的固定数据
np.random.seed(10)
data = np.random.normal(100, 20, 200)

# 2. 绘图
fig, ax = plt.subplots(figsize=(7, 4))

# 箱线图可以设置很多参数，比如颜色、离群点样式
box = ax.boxplot(
    data,
    patch_artist=True,  # 让箱子能填充颜色
    boxprops={'facecolor': '#87CEEB', 'edgecolor': 'black'},
    medianprops={'color': '#FF6700', 'linewidth': 2},  # 中位数线改成橙色，和你图里一致
    flierprops={'marker': 'o', 'markerfacecolor': 'white', 'markeredgecolor': 'black'}
)

ax.set_title("基础箱线图", fontsize=14)
ax.set_ylabel("数值分布")
ax.grid(alpha=0.3)  # 网格线透明度调低，不抢焦点

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 固定数据
np.random.seed(10)
data = np.random.normal(100, 20, 200)

# 计算四分位数、IQR、上下界
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 筛选离群点
outliers = data[(data < lower_bound) | (data > upper_bound)]

# 打印关键数值
print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"下界: {lower_bound:.2f}")
print(f"上界: {upper_bound:.2f}")
print(f"离群点数量: {len(outliers)} 个")
print(f"离群点:\n{np.round(outliers,2)}")

# 绘图
fig, ax = plt.subplots(figsize=(7, 4))

# 箱线图开启填充
bp = ax.boxplot(data, patch_artist=True)

# 美化箱子、中位数线
for box in bp['boxes']:
    box.set(facecolor='lightblue', edgecolor='black')
for med in bp['medians']:
    med.set(color='darkorange', linewidth=2)

# 手动标注每一个离群点
for val in outliers:
    ax.text(1.05, val, f"{val:.1f}", fontsize=8)

ax.set_title("箱线图（标注全部离群点）", fontsize=14)
ax.set_ylabel("数值分布")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
