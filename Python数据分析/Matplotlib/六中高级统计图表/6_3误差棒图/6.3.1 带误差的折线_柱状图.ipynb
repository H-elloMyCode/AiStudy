```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 4, 2, 5, 4]
```

5 组实验数据
每组都有一个测量值
每组都有一个误差范围（误差棒）
误差棒越长 = 数据波动越大、越不稳定

```python
error = [0.3, 0.4, 0.2, 0.5, 0.3] # 误差值：每个点的波动范围
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.errorbar(x, y, yerr=error, capsize=5, color='blue', linewidth=2)
ax1.set_title('误差折线图')
ax1.grid(alpha=0.3)

ax2.bar(x, y, yerr=error, capsize=5, color='orange')
ax2.set_title('误差柱状图')

plt.tight_layout()

plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.errorbar(x, y, yerr=error, capsize=10, color='blue', linewidth=2)
ax1.set_title('误差折线图')
ax1.grid(alpha=0.3)

ax2.bar(x, y, yerr=error, capsize=1, color='orange')
ax2.set_title('误差柱状图')

plt.tight_layout()

plt.show()
```
