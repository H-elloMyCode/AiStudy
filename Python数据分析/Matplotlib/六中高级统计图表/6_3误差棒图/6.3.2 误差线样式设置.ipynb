```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 1, 4, 2]
error = [0.2, 0.4, 0.3, 0.5, 0.2]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.errorbar(
    x,
    y,
    yerr=error,
    ecolor='crimson',
    capsize=20,
    elinewidth=10,
    capthick=2,
    color='navy',
    linewidth=2,
    marker='o',
    markersize=8
)

ax.set_title("误差线样式自定义", fontsize=14)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.errorbar(
    x,
    y,
    yerr=error,
    ecolor='crimson',
    capsize=15,
    elinewidth=2,
    capthick=5,
    color='navy',
    linewidth=2,
    marker='o',
    markersize=8
)

ax.set_title("误差线样式自定义", fontsize=14)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
