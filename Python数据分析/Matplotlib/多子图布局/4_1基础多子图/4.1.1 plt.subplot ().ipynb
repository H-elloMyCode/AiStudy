```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 2, 5, 4]
y2 = [2, 1, 4, 3, 1]
y3 = [3, 4, 1, 2, 3]
y4 = [5, 2, 3, 1, 4]
```

```python
plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.plot(x, y1, color='r')
plt.title('子图1')

plt.subplot(2, 2, 2)
plt.plot(x, y2, color='b')
plt.title('子图2')

plt.subplot(2, 2, 3)
plt.plot(x, y3, color='g')
plt.title('子图3')

plt.subplot(2, 2, 4)
plt.plot(x, y4, linestyle='--')
plt.title('子图4')


# plt.tight_layout()
plt.show()
```
