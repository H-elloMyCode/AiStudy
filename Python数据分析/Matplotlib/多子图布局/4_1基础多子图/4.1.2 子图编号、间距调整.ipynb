```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]
```

```python
plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title('子图1')

plt.subplot(2, 2, 2)
plt.bar(x, y)
plt.title('子图2')

plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title('子图3')

plt.subplot(2, 2, 4)
plt.plot(x, y, linestyle='--')
plt.title('子图4')

plt.tight_layout()

plt.show()
```
