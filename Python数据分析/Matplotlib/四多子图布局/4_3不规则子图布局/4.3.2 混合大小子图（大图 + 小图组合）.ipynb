```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
fig = plt.figure(figsize=(9, 5))

gs = gridspec.GridSpec(2, 2, figure=fig, wspace=0.3, hspace=0.3)

ax_big = fig.add_subplot(gs[:, 0])
ax_big.plot([1, 2, 3, 4, 5], [3, 1, 4, 2, 5], linewidth=2, color='crimson')
ax_big.set_title('左侧大图')

ax1 = fig.add_subplot(gs[0, 1])
ax1.bar([1, 2, 3], [2, 3, 1], color='skyblue')
ax1.set_title('右上小图')

ax2 = fig.add_subplot(gs[1, 1])
ax2.scatter([1, 2, 3], [1, 3, 2], color='orange')
ax2.set_title('右下小图')

plt.show()
```
