```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 4, 56, 5]
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
noise = np.random.normal(0, 0.2, 100)
```

```python
plt.figure(figsize=(7, 4))

plt.plot(x, y1, label="sin", linewidth=2)
plt.plot(x, y2, label="cos", linewidth=2)

plt.scatter(x, y1 + noise, alpha=0.5, s=20)

plt.title("NumPy 数组可视化", fontsize=14)
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
