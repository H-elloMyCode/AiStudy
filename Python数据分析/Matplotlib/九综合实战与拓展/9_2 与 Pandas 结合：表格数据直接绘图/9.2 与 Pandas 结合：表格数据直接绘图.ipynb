```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.DataFrame(
    {
        "年份": ["2020", "2021", "2022", "2023", "2024"],
        "收入": [12, 15, 18, 22, 28],
        "成本": [8, 9, 10, 12, 15]
    }
)

df
```

```python
df.plot(x='年份', kind='line', linewidth=2, figsize=(7, 4))

plt.title("Pandas 表格数据可视化", fontsize=14)
plt.grid(alpha=0.3)
```

```python
df.plot(x='年份', kind='line', linewidth=2, figsize=(7, 4))

plt.title("Pandas 表格数据可视化", fontsize=14)
plt.grid(alpha=0.3)

plt.savefig("pandas_plot.png", dpi=300, bbox_inches="tight")
plt.close()
print("Pandas 绘图完成！")
```
