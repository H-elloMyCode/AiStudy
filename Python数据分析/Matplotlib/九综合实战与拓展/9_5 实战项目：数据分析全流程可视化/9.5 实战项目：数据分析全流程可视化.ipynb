```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
```

```python
months = np.arange(1, 13)
print(months)
```

```python
revenue = np.random.randint(80, 150, 12)
revenue
```

```python
cost = revenue * np.random.uniform(0.5, 0.7, 12)
cost
```

```python
profit = revenue - cost
```

```python
profit
```

```python
df = pd.DataFrame({
    "月份": months,
    "收入": revenue,
    "成本": cost,
    "利润": profit
})

df
```

```python
fig = plt.figure(figsize=(12, 8))

ax1 = plt.subplot(2, 2, 1)
df.plot(x='月份', y=['收入', '成本'], ax=ax1, kind='line')
ax1.set_title('月度收入与成本')
ax1.grid(alpha=0.3)

ax2 = plt.subplot(2, 2, 2)
df.plot(x='月份', y='利润', ax=ax2, kind='bar', color='green')
ax2.set_title('月度利润')

ax3 = plt.subplot(2, 2, 3)
total = [df['收入'].sum(), df['成本'].sum()]
ax3.pie(total, labels=['总收入', '总成本'], autopct='%.1f%%', colors=['skyblue', 'orange'])
ax3.set_title("总收入成本占比")

ax4 = plt.subplot(2, 2, 4)
df['利润累计'] = df['利润'].cumsum()
df.plot(x='月份', y='利润累计', ax=ax4, color='red')
ax4.set_title('累计利润趋势')

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------
# OO 模式：创建画布 + 子图
# ----------------------
fig = plt.figure(figsize=(12, 8))  # 创建总画布

# 逐个创建子图（OO 模式）
ax1 = fig.add_subplot(2, 2, 1)  # 左上
ax2 = fig.add_subplot(2, 2, 2)  # 右上
ax3 = fig.add_subplot(2, 2, 3)  # 左下
ax4 = fig.add_subplot(2, 2, 4)  # 右下

# ----------------------
# 子图1：月度收入与成本（折线图）
# ----------------------
df.plot(x='月份', y=['收入', '成本'], ax=ax1, kind='line')
ax1.set_title('月度收入与成本')
ax1.grid(alpha=0.3)

# ----------------------
# 子图2：月度利润（柱状图）
# ----------------------
df.plot(x='月份', y='利润', ax=ax2, kind='bar', color='green')
ax2.set_title('月度利润')

# ----------------------
# 子图3：总收入成本占比（饼图）
# ----------------------
total = [df['收入'].sum(), df['成本'].sum()]
ax3.pie(total, labels=['总收入', '总成本'], autopct='%.1f%%', colors=['skyblue', 'orange'])
ax3.set_title("总收入成本占比")

# ----------------------
# 子图4：累计利润趋势
# ----------------------
df['利润累计'] = df['利润'].cumsum()
df.plot(x='月份', y='利润累计', ax=ax4, color='red')
ax4.set_title('累计利润趋势')

# ----------------------
# 布局调整 + 显示
# ----------------------
fig.tight_layout()  # OO 模式写法
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ✅ 标准 OO 模式：一句创建 fig + 2x2 子图
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# 拿到 4 个子图对象
ax1 = axs[0, 0]
ax2 = axs[0, 1]
ax3 = axs[1, 0]
ax4 = axs[1, 1]

# 子图1
df.plot(x='月份', y=['收入', '成本'], ax=ax1, kind='line')
ax1.set_title('月度收入与成本')
ax1.grid(alpha=0.3)

# 子图2
df.plot(x='月份', y='利润', ax=ax2, kind='bar', color='green')
ax2.set_title('月度利润')

# 子图3
total = [df['收入'].sum(), df['成本'].sum()]
ax3.pie(total, labels=['总收入', '总成本'], autopct='%.1f%%', colors=['skyblue','orange'])
ax3.set_title("总收入成本占比")

# 子图4
df['利润累计'] = df['利润'].cumsum()
df.plot(x='月份', y='利润累计', ax=ax4, color='red')
ax4.set_title('累计利润趋势')

fig.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 标准 OO 模式：创建 2x2 子图
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 8))

# ----------------------
# 子图1：折线图 —— 直接用 ax1.plot()
# ----------------------
ax1.plot(df['月份'], df['收入'], label='收入', linewidth=2)
ax1.plot(df['月份'], df['成本'], label='成本', linewidth=2)
ax1.set_title('月度收入与成本')
ax1.legend()  # 显示图例
ax1.grid(alpha=0.3)

# ----------------------
# 子图2：柱状图 —— 直接用 ax2.bar()
# ----------------------
ax2.bar(df['月份'], df['利润'], color='green')
ax2.set_title('月度利润')

# ----------------------
# 子图3：饼图 —— ax3.pie()
# ----------------------
total = [df['收入'].sum(), df['成本'].sum()]
ax3.pie(total, labels=['总收入', '总成本'], autopct='%.1f%%',
        colors=['skyblue', 'orange'])
ax3.set_title("总收入成本占比")

# ----------------------
# 子图4：累计利润折线图
# ----------------------
df['利润累计'] = df['利润'].cumsum()
ax4.plot(df['月份'], df['利润累计'], color='red', linewidth=2)
ax4.set_title('累计利润趋势')

# 调整布局
fig.tight_layout()
plt.show()
```
