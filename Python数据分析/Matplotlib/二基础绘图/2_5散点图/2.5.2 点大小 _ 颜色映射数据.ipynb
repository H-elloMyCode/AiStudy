```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

size = x * 200
color = y

fig, ax = plt.subplots(figsize=(7, 4))

sc = ax.scatter(
    x,
    y,
    s = size,
    c = color,
    cmap='viridis',
    alpha = 0.7
)

plt.colorbar(sc, label='颜色数据')

plt.show()
```
