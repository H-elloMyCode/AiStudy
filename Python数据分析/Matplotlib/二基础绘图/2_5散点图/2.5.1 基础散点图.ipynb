```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

fig, ax = plt.subplots(figsize=(7, 4))

ax.scatter(x, y, color='blue', alpha=0.7, s=50, edgecolor='black')

ax.set_title('基础散点图')
ax.set_xlabel('X变量')
ax.set_ylabel('Y变量')

plt.show()
```
