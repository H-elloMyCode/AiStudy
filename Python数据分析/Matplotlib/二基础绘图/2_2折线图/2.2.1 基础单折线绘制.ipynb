```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [3, 5, 2, 7, 4, 6]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o')

ax.set_title('基础单折线图')
ax.set_xlabel('X轴')
ax.set_ylabel('Y轴')

plt.show()
```
