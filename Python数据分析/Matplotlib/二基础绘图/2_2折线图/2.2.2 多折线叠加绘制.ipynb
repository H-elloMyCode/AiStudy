```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = [1, 2, 3, 4, 5, 6]
y1 = [2, 4, 1, 5, 3, 4]
y2 = [3, 1, 4, 2, 5, 2]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y1, marker='o', label='销售额')
ax.plot(x, y2, marker='s', label='利润')

ax.set_title('多折线叠加图')
ax.legend()

plt.show()
```
