```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 5, 2, 4, 1]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(
    x,
    y,
    color='darkblue',
    linestyle='-',
    linewidth=2,
    marker='*',
    markersize=10,
    label='自定义样式'
)

ax.set_title('折线样式设置')
ax.legend()

plt.show()
```
