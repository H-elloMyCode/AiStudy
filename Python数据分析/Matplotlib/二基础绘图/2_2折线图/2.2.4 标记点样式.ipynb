```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [4, 2, 5, 3, 6, 1]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(
    x,
    y,
    linewidth=2,
    marker='D',
    markersize=10,
    markerfacecolor='black',
    markeredgecolor='red',
    markeredgewidth=1
)

ax.set_title('标记点样式')

plt.show()
```
