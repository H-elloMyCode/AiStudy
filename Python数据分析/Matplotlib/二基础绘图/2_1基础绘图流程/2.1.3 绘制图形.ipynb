```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', label='数据曲线')

plt.show()
```
