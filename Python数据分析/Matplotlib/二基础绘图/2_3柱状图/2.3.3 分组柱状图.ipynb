```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
import numpy as np
```

```python
x = np.array([1, 2, 3, 4])

y1 = [20, 35, 30, 25]
y2 = [15, 30, 25, 20]

width = 0.35

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(x - width / 2, y1, width, label='上半年', color='skyblue')
ax.bar(x + width / 2, y2, width, label='下半年', color='orange')

ax.set_title('分组柱状图')
ax.set_xticks(x)
ax.set_xticklabels(['苹果', '香蕉', '橙子', '葡萄'])
ax.set_xlabel('水果分类')
ax.legend()

plt.show()
```

```python
x = np.array([1, 2, 3, 4])

y1 = [20, 35, 30, 25]
y2 = [15, 30, 25, 20]
y3 = [14, 21, 5, 2]

width = 0.25

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(x - width, y1, width, label='上半年', color='skyblue')
ax.bar(x, y2, width, label='下半年', color='orange')
ax.bar(x + width, y2, width, label='未知', color='green')

ax.set_title('分组柱状图')
ax.set_xticks(x)
ax.set_xticklabels(['苹果', '香蕉', '橙子', '葡萄'])
ax.set_xlabel('水果分类')
ax.legend()

plt.show()
```
