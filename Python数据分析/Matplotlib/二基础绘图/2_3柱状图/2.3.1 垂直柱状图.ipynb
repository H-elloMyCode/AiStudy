```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = ['苹果', '香蕉', '橙子', '葡萄']
y = [15, 22, 18, 25]

fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(x, y, color='skyblue', width=0.5, edgecolor='black')

ax.set_title('水果销量统计')
ax.set_xlabel('水果种类')
ax.set_ylabel('销量')

plt.show()
```
