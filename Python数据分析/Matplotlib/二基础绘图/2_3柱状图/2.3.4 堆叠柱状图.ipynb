```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = ['苹果', '香蕉', '橙子', '葡萄']

y1 = [10, 15, 12, 18]
y2 = [5, 7, 6, 7]

fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(x, y1, label='线下', color='skyblue')
ax.bar(x, y2, bottom=y1, label='线上', color='orange')

ax.set_title('堆叠柱状图(总销量)')
ax.set_ylabel('销量')
ax.legend()

plt.show()
```
