```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
labels = ['苹果', '香蕉', '橙子', '葡萄']
sizes = [35, 25, 20, 20]

fig, ax = plt.subplots(figsize=(6, 6))

ax.pie(
    sizes,
    labels=labels,
    autopct='%.1f%%',
    startangle=90
)

ax.set_title('水果销量占比饼图')
ax.axis('equal')

plt.show()
```
