```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
sizes = [35, 25, 20, 20]
labels = ['苹果', '香蕉', '橙子', '葡萄']

explode = [0.1, 0, 0, 0]

fig, ax = plt.subplots(figsize=(7, 6))

ax.pie(
    sizes,
    explode=explode,
    labels=labels,
    autopct='%1.1f%%',
    shadow=True,
    startangle=90
)

ax.legend(labels, loc='best')
ax.axis('equal')
ax.set_title('水果销量占比')

plt.show()
```
