```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = np.linspace(0, 100, 100000000)
y = np.sin(x) + np.random.normal(0, 0.1, len(x))

len(x)
```

```python
len(y)
```

```python
plt.figure(figsize=(8, 4))

plt.plot(x, y)

plt.title("百万数据优化绘图", fontsize=14)

plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.plot(x[::10], y[::10], linewidth=1, color='blue')

plt.title("百万数据优化绘图", fontsize=14)

plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.plot(x[::10], y[::10], linewidth=1, color='blue')

plt.title("百万数据优化绘图", fontsize=14)
plt.savefig("大数据优化图.png", dpi=150, bbox_inches='tight')

plt.close()

print("大数据绘图优化完成！")
```

```python
plt.figure(figsize=(8, 4))

plt.plot(x, y, linewidth=1, marker='', antialiased=False)

plt.title("百万数据优化绘图", fontsize=14)
plt.savefig("大数据优化图1.png", dpi=150, bbox_inches='tight')

plt.close()

print("大数据绘图优化完成！")
```
