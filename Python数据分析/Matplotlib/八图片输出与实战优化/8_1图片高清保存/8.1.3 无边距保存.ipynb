```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x, y, linewidth=2)

plt.savefig(
    "无边距图.png",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0,
    facecolor="white"
)

# plt.close()
plt.show()
print("无边距保存成功!")
```
