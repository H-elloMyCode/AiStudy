```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = np.arange(10)
x
```

```python
y = x ** 2
y
```

```python
plt.figure(figsize=(6, 4))
plt.plot(x, y, linewidth=3, color='blue')
plt.title("高清保存测试 dpi=300")

plt.savefig("高清图.png", dpi=300, bbox_inches='tight')

plt.close()
print("保存成功!")
```

```python
plt.figure(figsize=(6, 4))
plt.plot(x, y, linewidth=3, color='blue')
plt.title("高清保存测试 dpi=300")

plt.savefig("超高清图.png", dpi=600, bbox_inches='tight')

plt.close()
print("保存成功!")
```
