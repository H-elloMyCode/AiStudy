```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# ==================== 解决中文乱码（全局生效）====================
plt.rcParams["font.sans-serif"] = ["SimHei", "Microsoft YaHei", "Arial Unicode MS"]
# plt.rcParams["axes.unicode_minus"] = False  # 解决负号 - 变方框问题
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 4, 1]
```

```python
plt.figure(figsize=(6, 4))

plt.plot(x, y)
plt.title("中文标题测试", fontsize=14)
plt.xlabel("X轴 中文", fontsize=12)
plt.ylabel("Y轴 中文", fontsize=12)
plt.grid(alpha=0.3)

# plt.savefig("中文测试.png", dpi=300, bbox_inches='tight')

# plt.close()
plt.show()
```
