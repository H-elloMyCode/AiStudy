```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [-3, -2, -1, 0, 1, 2, 3]
y = [-2, -1, 0, 1, 2, 3, 4]
```

```python
plt.figure(figsize=(6, 4))

plt.plot(x, y, linewidth=2)
plt.title("负号正常显示测试", fontsize=14)

# plt.show()
# 保存
plt.savefig("负号修复.png", dpi=300, bbox_inches='tight')
plt.close()

print("负号修复成功！")
```
