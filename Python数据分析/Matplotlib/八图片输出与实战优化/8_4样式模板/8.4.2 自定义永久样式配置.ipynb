```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams.update(
    {
        "font.sans-serif": ["SimHei", "Arial Unicode MS"],
        "axes.unicode_minus": False,
        "figure.figsize": [10, 6],
        "lines.linewidth": 5,
        "axes.titlesize": 20,
        "axes.grid": True,
        "grid.alpha": 0.3
    }
)
```

```python
x = np.linspace(0, 5, 20)
y = np.sin(x)

plt.plot(x, y)
plt.title("永久自定义样式测试")
plt.xlabel("X轴")
plt.ylabel("Y轴")

plt.show()
```

```python
x = np.linspace(0, 5, 20)
y = np.cos(x)

plt.plot(x, y)
plt.title("永久自定义样式测试")
plt.xlabel("X轴")
plt.ylabel("Y轴")

plt.show()
```
