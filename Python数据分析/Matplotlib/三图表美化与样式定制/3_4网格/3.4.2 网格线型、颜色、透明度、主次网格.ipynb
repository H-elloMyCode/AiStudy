```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]

```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o')

ax.grid(
    True,
    which='major',
    linestyle='-',
    color='red',
    alpha=0.7
)

ax.grid(
    True,
    which='minor',
    linestyle=':',
    color='blue',
    alpha=0.4
)

ax.minorticks_on()

ax.set_title("网格样式+主次网格", fontsize=14)

plt.show()
```
