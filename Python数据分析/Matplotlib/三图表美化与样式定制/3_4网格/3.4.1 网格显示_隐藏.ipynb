```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.grid()

ax.set_title("网格显示与隐藏", fontsize=14)

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.grid(True)

ax.set_title("网格显示与隐藏", fontsize=14)

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.grid(False)

ax.set_title("网格显示与隐藏", fontsize=14)

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.grid(True, axis='x')

ax.set_title("网格显示与隐藏", fontsize=14)

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.grid(True, axis='y')

ax.set_title("网格显示与隐藏", fontsize=14)

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.grid(True, axis='both')

ax.set_title("网格显示与隐藏", fontsize=14)

plt.show()
```
