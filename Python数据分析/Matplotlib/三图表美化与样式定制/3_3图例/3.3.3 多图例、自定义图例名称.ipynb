```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 1, 5, 3]
y2 = [1, 3, 5, 2, 4]
y3 = [3, 1, 2, 5, 4]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

line1, = ax.plot(x, y1, color='red', label='商品A销量')
line2, = ax.plot(x, y2, color='blue', label='商品B销量')
line3, = ax.plot(x, y3, color='green', label='总成本')

legend1 = ax.legend(handles=[line1, line2], loc='upper left', fontsize=10)
ax.add_artist(legend1)

ax.legend(handles=[line3], loc='lower right', fontsize=10)

ax.set_title("多图例 + 自定义名称", fontsize=14)
plt.show()
```
