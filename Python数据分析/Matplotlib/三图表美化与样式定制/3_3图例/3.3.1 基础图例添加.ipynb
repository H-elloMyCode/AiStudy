```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 1, 5, 3]
y2 = [1, 3, 5, 2, 4]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y1, label='销量')
ax.plot(x, y2, label='库存')

ax.legend()

# 标题与标签
ax.set_title("基础图例演示", fontsize=14)
ax.set_xlabel("X轴")
ax.set_ylabel("Y轴")

plt.show()
```
