```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o')

ax.set_title(
    '月度数据变化趋势',
    fontsize=16,
    fontweight='bold',
    color='darkred',
    loc='center',
    pad=15
)

plt.show()
```
