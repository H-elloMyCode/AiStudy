```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, y)

ax.set_title(
    'Demo',
    fontsize=18,
    color='darkred',
    fontweight='bold',
    fontstyle='italic',
    loc='center',
    pad=20
)

plt.show()
```
