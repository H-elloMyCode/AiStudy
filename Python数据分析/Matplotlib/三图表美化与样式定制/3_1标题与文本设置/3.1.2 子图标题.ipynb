```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = np.arange(1, 6)
x
```

```python
y1 = [1, 3, 2, 5, 4]
y2 = [2, 1, 4, 3, 5]
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.plot(x, y1)
ax1.set_title(
    'Sale',
    fontsize=14,
    fontweight='bold',
    color='darkblue'
)

ax2.bar(x, y2)
ax2.set_title(
    'Inventory',
    fontsize=14,
    fontweight='bold',
    color='darkred'
)

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.plot(x, y1)
ax1.set_title(
    '销量数据',
    fontsize=14,
    fontweight='bold',
    color='darkblue'
)

ax2.bar(x, y2)
ax2.set_title(
    '库存数据',
    fontsize=14,
    fontweight='bold',
    color='darkred'
)

plt.tight_layout()
plt.show()
```
