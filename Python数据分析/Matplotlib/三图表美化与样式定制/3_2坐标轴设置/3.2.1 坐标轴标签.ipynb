```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o')

ax.set_xlabel('月份', fontsize=12, fontweight='bold', color='blue', labelpad=8)
ax.set_ylabel('销量(万元)', fontsize=12, fontweight='bold', color='red', labelpad=8)

ax.set_title('月度销量趋势', fontsize=16, pad=15)

plt.show()
```
