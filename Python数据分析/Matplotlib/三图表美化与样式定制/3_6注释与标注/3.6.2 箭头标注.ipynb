```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.annotate(
    "最高值",
    xy=(5, 5),
    xytext=(3.5, 4.5),
    fontsize=12,
    color='red',
    fontweight='bold',
    arrowprops=dict(arrowstyle="fancy", color='red')
    # arrowprops=dict(arrowstyle="fancy,head_width=0.6,head_length=0.9", color='red')
    # arrowprops=dict(arrowstyle="round,head_length=1.2", color='red')
)

ax.set_title("箭头标注演示", fontsize=14)
ax.grid(alpha=0.3)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.annotate(
    "height",
    xy=(5, 5),
    xytext=(3.5, 4.5),
    fontsize=12,
    color='red',
    fontweight='bold',
    arrowprops=dict(arrowstyle='->', color='red')
)

ax.set_title("箭头标注演示", fontsize=14)
ax.grid(alpha=0.3)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.annotate(
    "谷点",
    xy=(4, 2),
    xytext=(3.8, 3.0),
    fontsize=12,
    color='red',
    fontweight='bold',
    arrowprops=dict(arrowstyle='fancy', color='green')
)

ax.set_title("箭头标注演示", fontsize=14)
ax.grid(alpha=0.3)
plt.show()
```
