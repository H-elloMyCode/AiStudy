```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2)

ax.text(
    3,
    4,
    "text",
    color='red',
    fontweight='bold',
    bbox=dict(boxstyle="sawtooth", facecolor="yellow", alpha=0.5)
)

ax.set_title("文本标注演示", fontsize=14)
ax.grid(alpha=0.3)

plt.show()
```
