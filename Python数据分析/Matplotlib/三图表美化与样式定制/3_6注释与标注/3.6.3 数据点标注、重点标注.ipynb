```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, marker='o', linewidth=2, color='#3366ff')

for xi, yi in zip(x, y):
    ax.text(xi, yi + 0.3, f'{yi}', fontsize=11, ha='center', fontweight='bold')

ax.scatter(5, 5, color='red', s=100, zorder=5)
ax.text(4.7, 5.2, '最大值', color='red', fontsize=12, ha='center')

ax.set_title("数据点+重点标注", fontsize=14)
ax.set_ylim(0, 6)
ax.grid(alpha=0.3)
plt.show()
```
