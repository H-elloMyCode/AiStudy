```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 2, 5, 4]
y2 = [2, 1, 4, 3, 1]
y3 = [3, 4, 1, 2, 3]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

custom_colors = ["#FF5733", "#33A1FF", "#33FF57"]

ax.plot(x, y1, color=custom_colors[0], linewidth=2, label='系列1')
ax.plot(x, y2, color=custom_colors[1], linewidth=2, label='系列2')
ax.plot(x, y3, color=custom_colors[2], linewidth=2, label='系列3')

ax.legend()
ax.set_title("内置配色 + 自定义配色", fontsize=14)
ax.grid(alpha=0.3)

plt.show()
```
