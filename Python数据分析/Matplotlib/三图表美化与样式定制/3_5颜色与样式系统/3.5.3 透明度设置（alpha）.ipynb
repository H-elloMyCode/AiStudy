```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y1 = [3, 1, 4, 2, 5]
y2 = [2, 4, 1, 5, 3]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(x, y1, color='blue', label='数据1')
ax.bar(x, y2, color='red', label='数据2')

ax.grid(True, alpha=0.3)

ax.legend()

ax.set_title("透明度 alpha 设置", fontsize=14)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(x, y1, color='blue', label='数据1')
ax.bar(x, y2, color='red', label='数据2')

ax.grid(True)

ax.legend()

ax.set_title("透明度 alpha 设置", fontsize=14)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(x, y1, color='blue', alpha=0.4, label='数据1')
ax.bar(x, y2, color='red', alpha=0.4, label='数据2')

ax.grid(True, alpha=0.3)

ax.legend(facecolor="black", framealpha=0.9)

ax.set_title("透明度 alpha 设置", fontsize=14)
plt.show()
```
