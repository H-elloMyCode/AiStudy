```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(
    x,
    y,
    color='#FF5733',
    linewidth=3
)

ax.set_title("颜色表示法演示", fontsize=14)
ax.grid(True, alpha=0.3)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(
    x,
    y,
    # color='#FF5733',
    color='red',
    linewidth=3
)

ax.set_title("颜色表示法演示", fontsize=14)
ax.grid(True, alpha=0.3)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(
    x,
    y,
    # color='#FF5733',
    # color='red',
    color=(1, 0, 0),
    linewidth=3
)

ax.set_title("颜色表示法演示", fontsize=14)
ax.grid(True, alpha=0.3)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(
    x,
    y,
    # color='#FF5733',
    # color='red',
    # color=(1, 0, 0),
    color='r',
    linewidth=3
)

ax.set_title("颜色表示法演示", fontsize=14)
ax.grid(True, alpha=0.3)
plt.show()
```
