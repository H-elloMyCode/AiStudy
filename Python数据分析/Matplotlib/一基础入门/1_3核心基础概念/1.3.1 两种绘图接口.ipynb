```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
x = [1, 2, 3, 4]
y = [2, 4, 1, 5]
```

```python
plt.plot(x, y)

plt.title('pyplot 接口')
plt.xlabel('X')
plt.ylabel('Y')

plt.show()
```

```python
x = [1, 2, 3, 4]
y = [2, 4, 1, 5]
```

```python
fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title('面向对象接口')
ax.set_xlabel('X')
ax.set_ylabel('Y')

plt.show()
```
