```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 4]

ax.plot(x, y, linewidth=2, label='销量趋势')

ax.set_title('Matplotlib 完整图表元素', fontsize=12)
ax.set_xlabel("X轴（月份）")
ax.set_ylabel("Y轴（销售额）")

ax.set_xlim(0, 6)
ax.set_ylim(0, 6)
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```
