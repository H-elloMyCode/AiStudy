```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.sans-serif'] = ['SimHei']  # 黑体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot([1, 2, 3, 4], [3, 5, 2, 6])
ax.set_xlabel('X 轴')
ax.set_ylabel('Y 轴')

ax.set_xlim(0, 5)
ax.set_ylim(0, 8)

ax.set_title('坐标轴演示')

plt.show()
```
