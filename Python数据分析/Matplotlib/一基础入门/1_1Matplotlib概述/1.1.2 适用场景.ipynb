```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df = pd.read_csv('scene_data.csv')

# 创建画布 + 2行1列的子图（上下布局）
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# 第一个子图：销售额 折线图
ax1.plot(df['月份'], df['销售额'], marker='o', color='#2E86AB', linewidth=2, label='销售额')
ax1.set_title('月度销售额趋势', fontsize=10)
ax1.set_xlabel('月份')
ax1.set_ylabel('销售额')
ax1.legend()
ax1.grid(alpha=0.3)
# 如果需要固定Y轴 0-100，打开下面这句
# ax1.set_ylim(0, 100)

# 第二个子图：订单量 柱状图
ax2.bar(df['月份'], df['订单量'], color='#A23B72', alpha=0.7, label='订单量')
ax2.set_title('月度订单量统计', fontsize=10)
ax2.set_xlabel('月份')
ax2.set_ylabel('订单量')
ax2.legend()
ax2.grid(alpha=0.3, axis='y')

# 整体标题
fig.suptitle('月度销售业务数据展示', fontsize=12, y=0.98)

plt.tight_layout()  # 自动调整间距
plt.show()
```
