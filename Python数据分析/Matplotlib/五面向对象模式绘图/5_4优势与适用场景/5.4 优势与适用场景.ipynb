```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 3, 2, 5, 4]
y2 = [2, 1, 4, 3, 1]
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 3))

ax1.plot(x, y1)
ax1.set_title('子图1')

ax2.plot(x, y2)
ax2.set_title('子图2')

plt.tight_layout()
plt.show()
```
