```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

# ax.plot(x, y, color='red', linewidth=2, label='折线图')
# ax.bar(x, y, color='skyblue', label='柱状图')
ax.scatter(x, y, color='green', s=80, label='散点图')

ax.set_title("OO 模式基础图表", fontsize=14)
ax.set_xlabel("X轴")
ax.set_ylabel("Y轴")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```
