```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 1, 4, 2, 5]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(x, y, 'o-', label='数据曲线')

# 标题
ax.set_title('OO 模式图表元素设置', fontsize=14, pad=10)

# 坐标轴标签
ax.set_xlabel('X 轴名称', fontsize=12)
ax.set_ylabel('Y 轴数值', fontsize=12)

# 坐标轴范围
ax.set_xlim(0, 6)
ax.set_ylim(0, 6)

# 图例
ax.legend(loc='upper left')
# 网格
ax.grid(True, alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```
