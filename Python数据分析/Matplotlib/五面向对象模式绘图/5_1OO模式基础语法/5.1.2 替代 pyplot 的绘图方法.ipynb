```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

ax.plot(x, y, linewidth=2, color='#FF6B6B', label='数据线')

ax.set_title("OO 模式完整示例", fontsize=14)
ax.set_xlabel("X 轴")
ax.set_ylabel("Y 轴")
ax.grid(alpha=0.3)
ax.legend()

plt.tight_layout()

# plt.set_title()

plt.show()
```

```python
# figure = plt.(figsize=(7, 4))
plt.figure(figsize=(7, 4))

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

plt.plot(x, y, linewidth=2, color='#FF6B6B', label='数据线')

plt.title("传统模式示例", fontsize=14)
plt.xlabel("X 轴")
plt.ylabel("Y 轴")
plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()

plt.bar(x, y)

# plt.set_title()

plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(7, 4))

x = [1, 2, 3, 4, 5]
y1 = [2, 1, 1, 10, 3]
y2 = [2, 4, 2, 5, 4]

ax1.plot(x, y1, linewidth=2, color='#FF6B6B', label='数据线')

ax1.set_title("OO 模式完整示例", fontsize=14)
ax1.set_xlabel("X 轴")
ax1.set_ylabel("Y 轴")
ax1.grid(alpha=0.3)
ax1.legend()

ax2.plot(x, y2)

plt.tight_layout()

# plt.set_title()

plt.show()
```
