```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot([1, 2, 3, 4], [2, 3, 1, 4], linewidth=2, color='crimson')

ax.set_title('OO 模式绘图', fontsize=14)
ax.set_xlabel('X 轴')
ax.set_ylabel('Y 轴')

ax.grid(alpha=0.3)


plt.tight_layout()
plt.show()
```
