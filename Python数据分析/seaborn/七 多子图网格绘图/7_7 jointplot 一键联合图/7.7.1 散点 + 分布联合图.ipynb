```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",

    kind="scatter",
    marginal_kws={"color": "teal", "alpha": 0.6},

    color="orange"
)

plt.suptitle("jointplot 散点+分布联合图", y=1.02)
plt.tight_layout()
plt.show()
```
