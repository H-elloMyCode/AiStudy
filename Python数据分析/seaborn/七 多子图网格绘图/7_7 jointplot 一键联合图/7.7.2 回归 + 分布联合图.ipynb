```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",

    kind="reg", ## 导致直方图添加了 kde 曲线
    marginal_kws={"alpha": 0.6},

    color="#2E86AB"
)

plt.suptitle("jointplot 回归+分布联合图", y=1.02)
plt.tight_layout()
plt.show()
```
