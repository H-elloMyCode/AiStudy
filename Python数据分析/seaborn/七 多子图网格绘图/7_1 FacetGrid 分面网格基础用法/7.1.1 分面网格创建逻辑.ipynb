```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.FacetGrid(
    tips,
    col="time",
    height=4,
    aspect=0.8
)

g.map(plt.scatter, "total_bill", "tip", color="#2E86AB")

g.set_titles(col_template="{col_name}")

plt.suptitle("FacetGrid 分面网格基础用法", y=1.05)
plt.tight_layout()

plt.show()
```
