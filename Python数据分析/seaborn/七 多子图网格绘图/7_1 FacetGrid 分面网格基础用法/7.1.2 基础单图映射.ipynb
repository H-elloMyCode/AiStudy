```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 中文适配
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.FacetGrid(
    tips,
    col="sex",
    height=4
)

g.map(sns.boxplot, "day", "total_bill", order=["Thur", "Fri", "Sat", "Sun"])

# 标题美化
g.set_titles("{col_name}")
plt.tight_layout()

plt.show()
```

```python
g = sns.FacetGrid(
    tips,
    row="sex",
    height=4
)

g.map(sns.boxplot, "day", "total_bill", order=["Thur", "Fri", "Sat", "Sun"])

# 标题美化
g.set_titles("{row_name}")
plt.tight_layout()

plt.show()
```
