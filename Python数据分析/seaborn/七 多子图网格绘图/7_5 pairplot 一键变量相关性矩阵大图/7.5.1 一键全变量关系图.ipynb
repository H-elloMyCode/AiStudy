```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
sns.pairplot(tips, kind="scatter", diag_kind="hist", palette="coolwarm")

plt.suptitle("pairplot 一键全变量关系大图", y=1.02)
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(tips, kind="scatter", diag_kind="hist", hue="smoker", palette="Blues")

plt.suptitle("pairplot 一键全变量关系大图", y=1.02)
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(tips, kind="scatter", diag_kind="hist", hue="smoker", palette="coolwarm")

plt.suptitle("pairplot 一键全变量关系大图", y=1.02)
plt.tight_layout()
plt.show()
```
