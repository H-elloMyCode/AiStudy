```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
sns.pairplot(
    tips,
    hue="time",

    diag_kind="kde",
    palette="Set2",

    kind="scatter"
)

plt.suptitle("pairplot 分组着色 & 对角图修改", y=1.02)
plt.tight_layout()
plt.show()
```

```python
sns.pairplot(
    tips,
    hue="time",

    diag_kind="kde",
    palette="Greens",

    kind="scatter"
)

plt.suptitle("pairplot 分组着色 & 对角图修改", y=1.02)
plt.tight_layout()
plt.show()
```
