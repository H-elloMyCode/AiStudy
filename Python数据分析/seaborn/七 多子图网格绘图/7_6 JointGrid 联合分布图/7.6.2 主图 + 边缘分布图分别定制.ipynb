```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.JointGrid(x="total_bill", y="tip", data=tips, height=6)

g.plot_joint(sns.regplot, color="#2E86AB", scatter_kws={"alpha": 0.6})

sns.kdeplot(x=tips["total_bill"], ax=g.ax_marg_x, color="red", fill=True)

sns.histplot(y=tips["tip"], ax=g.ax_marg_y, color="green", alpha=0.5)

plt.suptitle("JointGrid 主图与边缘图分别定制", y=1.03)
plt.tight_layout()
plt.show()
```
