```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.JointGrid(x="total_bill", y="tip", data=tips, height=6)

g.plot_joint(sns.scatterplot, color="teal")

g.plot_marginals(sns.histplot, color="orange", alpha=0.5)

plt.suptitle("JointGrid 自定义联合画布", y=1.03)
plt.tight_layout()
plt.show()
```
