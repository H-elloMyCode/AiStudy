```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.FacetGrid(
    tips,
    col="day",
    height=4,
    aspect=0.7
)

g.map(sns.scatterplot, "total_bill", "tip", color="teal")

g.set_titles("{col_name}")
plt.suptitle("按列拆分多子图", y=1.05)
plt.tight_layout()
plt.show()
```
