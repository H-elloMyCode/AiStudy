```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.FacetGrid(
    tips,
    row="day",
    height=3,
    aspect=2
)

g.map(sns.scatterplot, "total_bill", "tip", color="orange")

g.set_titles("{row_name}")
plt.suptitle("按行拆分多子图", y=1.05)
plt.tight_layout()
plt.show()
```
