```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
def my_plot(x, y, **kwargs):
    sns.regplot(x=x, y=y, line_kws={"color": "red"}, scatter_kws={"alpha": 0.6})
```

```python
g = sns.FacetGrid(tips, col="time", height=4, aspect=0.8)

g.map(my_plot, "total_bill", "tip")

# g.map(sns.boxplot, "", "")

g.set_titles("{col_name}")
plt.suptitle("自定义函数绑定分面网格", y=1.05)
plt.tight_layout()
plt.show()
```
