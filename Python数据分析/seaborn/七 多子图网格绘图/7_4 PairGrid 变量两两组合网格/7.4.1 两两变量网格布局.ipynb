```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.PairGrid(tips, height=2)

g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)

plt.suptitle("PairGrid 两两变量网格布局", y=1.02)
plt.tight_layout()
plt.show()
```
