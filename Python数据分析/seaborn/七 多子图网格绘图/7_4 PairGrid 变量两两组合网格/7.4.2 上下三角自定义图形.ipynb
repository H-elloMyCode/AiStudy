```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
g = sns.PairGrid(tips)

g.map_diag(sns.histplot, color="skyblue")
g.map_upper(sns.scatterplot, color="purple", alpha=0.6)
g.map_lower(sns.kdeplot, cmap="Blues", fill=True)

plt.suptitle("PairGrid 上下三角自定义图形", y=1.02)
plt.tight_layout()
plt.show()
```
