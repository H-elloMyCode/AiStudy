```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(7, 4))

sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    inner="box"
)

plt.title("小提琴图：内部样式切换（四分位线）")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    inner="quartile"
)

plt.title("小提琴图：内部样式切换（四分位线）")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    inner="point"
)

plt.title("小提琴图：内部样式切换（四分位线）")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    inner="stick"
)

plt.title("小提琴图：内部样式切换（四分位线）")
plt.tight_layout()
plt.show()
```
