```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(7, 4))

sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("二维嵌套分类图：日期 × 用餐时间")
# plt.legend(title="时段")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="time"
)

plt.title("二维嵌套分类图：日期 × 用餐时间")
# plt.legend(title="时段")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="time"
)

plt.title("二维嵌套分类图：日期 × 用餐时间")
plt.legend(title="时段")
plt.tight_layout()
plt.show()
```
