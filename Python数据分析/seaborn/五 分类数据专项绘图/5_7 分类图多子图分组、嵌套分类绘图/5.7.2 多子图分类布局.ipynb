```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    kind="box",
    col="time",
    height=4,
    aspect=0.8
)

plt.suptitle("多子图分类布局（午餐 vs 晚餐）", y=1.05)
plt.show()
```

```python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    kind="box",
    col="sex",
    height=4,
    aspect=0.8
)

plt.suptitle("多子图分类布局（男 vs 女）", y=1.05)
plt.show()
```
