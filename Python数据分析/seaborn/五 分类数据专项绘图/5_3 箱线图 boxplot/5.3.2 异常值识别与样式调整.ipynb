```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("箱线图异常值标记 + 样式自定义")
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    flierprops={"marker": "o", "markerfacecolor": "red", "markersize": 6, "alpha": 0.7}
)

plt.title("箱线图异常值标记 + 样式自定义")
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    flierprops={"marker": "o", "markerfacecolor": "red", "markersize": 6, "alpha": 0.7},
    boxprops={"facecolor": "lightblue", "alpha": 0.6}
)

plt.title("箱线图异常值标记 + 样式自定义")
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    flierprops={"marker": "o", "markerfacecolor": "red", "markersize": 6, "alpha": 0.7},
    boxprops={"facecolor": "lightblue", "alpha": 0.6},
    medianprops={"color": "darkblue", "linewidth": 2}
)

plt.title("箱线图异常值标记 + 样式自定义")
plt.show()
```
