```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(6, 4))

sns.barplot(data=tips, x="day", y="total_bill", ci=95)
plt.title("显示误差线（默认）")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.barplot(data=tips, x="day", y="total_bill", errorbar=('ci', 95))
plt.title("显示误差线（默认）")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.barplot(data=tips, x="day", y="total_bill", ci=None)
plt.title("关闭误差线")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.barplot(data=tips, x="day", y="total_bill", errorbar=None)
plt.title("关闭误差线")

plt.show()
```
