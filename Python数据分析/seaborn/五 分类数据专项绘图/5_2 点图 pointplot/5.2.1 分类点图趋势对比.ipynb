```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize = (7, 4))

sns.pointplot(
    data=tips,
    x="day",
    y="total_bill",
    color="#2E86AB"
)

plt.title("每日平均消费趋势（点图）")
plt.show()
```

```python
plt.figure(figsize = (7, 4))

sns.pointplot(
    data=tips,
    x="day",
    y="total_bill",
    color="#2E86AB",
    ci=None
)

plt.title("每日平均消费趋势（点图）")
plt.show()
```

```python
plt.figure(figsize = (7, 4))

sns.pointplot(
    data=tips,
    x="day",
    y="total_bill",
    color="#2E86AB",
    errorbar=None
)

plt.title("每日平均消费趋势（点图）")
plt.show()
```
