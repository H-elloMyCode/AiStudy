```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
df = pd.crosstab(tips["day"] , tips["sex"])

df
```

```python
df.plot(
    kind="bar",
    color=["#ff7f7f", "#7fbfef"],
    alpha=0.8
)

plt.title("堆叠条形图")
plt.xlabel("星期")
plt.ylabel("人数")
plt.legend(title="性别")
plt.tight_layout()
plt.show()
```

```python
df.plot(
    kind="bar",
    stacked=True,
    color=["#ff7f7f", "#7fbfef"],
    alpha=0.8
)

plt.title("堆叠条形图")
plt.xlabel("星期")
plt.ylabel("人数")
plt.legend(title="性别")
plt.tight_layout()
plt.show()
```

```python
# 提取数据、标签
x = np.arange(len(df.index))
width = 0.5
colors = ["#ff7f7f", "#7fbfef"]
alpha = 0.8

# 第一层：Female
plt.bar(x, df.iloc[:, 0], width, color=colors[0], alpha=alpha, label=df.columns[0])
# 第二层：Male，用 bottom 实现堆叠
plt.bar(x, df.iloc[:, 1], width, bottom=df.iloc[:, 0], color=colors[1], alpha=alpha, label=df.columns[1])

# 图表设置
plt.title("堆叠条形图")
plt.xlabel("星期")
plt.ylabel("人数")
plt.xticks(x, df.index)  # 设置x轴刻度为星期
plt.legend(title="性别")
plt.tight_layout()
plt.show()
```
