```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("多组分组条形图")
plt.legend(title="是否吸烟")
plt.tight_layout()
plt.show()
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker"
)

plt.title("多组分组条形图")
plt.legend(title="是否吸烟")
plt.tight_layout()
plt.show()
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    palette="viridis"
)

plt.title("多组分组条形图")
plt.legend(title="是否吸烟")
plt.tight_layout()
plt.show()
```
