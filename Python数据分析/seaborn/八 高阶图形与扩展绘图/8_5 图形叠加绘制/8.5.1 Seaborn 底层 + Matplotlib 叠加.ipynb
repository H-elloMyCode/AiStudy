```python
import seaborn as sns
import matplotlib.pyplot as plt
from pyparsing import line_end
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", color="#2E86AB")

plt.title("Seaborn + Matplotlib 叠加绘图")
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", color="#2E86AB")

plt.axhline(y=3, color="red", linestyle="--", linewidth=2)

plt.title("Seaborn + Matplotlib 叠加绘图")
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", color="#2E86AB")

plt.axhline(y=3, color="red", linestyle="--", linewidth=2)

plt.axvline(x=20, color="orange", linestyle=":")

plt.title("Seaborn + Matplotlib 叠加绘图")
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", color="#2E86AB")

plt.axhline(y=3, color="red", linestyle="--", linewidth=2)

plt.axvline(x=20, color="orange", linestyle=":")

plt.text(5, 8, "小费基准线", fontsize=10, color="red")

plt.title("Seaborn + Matplotlib 叠加绘图")
plt.tight_layout()
plt.show()
```
