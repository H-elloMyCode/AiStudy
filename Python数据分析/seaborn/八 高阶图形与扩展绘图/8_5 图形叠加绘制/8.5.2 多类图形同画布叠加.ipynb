```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.DataFrame(
    {
        "月份": [1, 2, 3, 4, 5, 6],
        "销量": [15, 28, 20, 35, 27, 40],
        "增长率": [3, 7, -2, 9, -1, 5]
    }
)

df
```

```python
plt.figure(figsize=(8, 4))

sns.barplot(data=df, x="月份", y="销量", color="#8dd3c7", alpha=0.7)

plt.title("多类图形同画布叠加")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.barplot(data=df, x="月份", y="销量", color="#8dd3c7", alpha=0.7)

sns.lineplot(data=df, x="月份", y="增长率", color="#fc8d62", marker="o", linewidth=3)

plt.title("多类图形同画布叠加")
plt.tight_layout()
plt.show()
```
