```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
x = np.arange(1, 13)
x
```

```python
y = np.random.randint(20, 50, size=12)

y
```

```python
plt.figure(figsize=(9, 4))

sns.lineplot(x=x, y=y, color="#2E86AB", linewidth=2)

plt.title("基础面积图")
plt.xlabel("序号")
plt.ylabel("数值")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4))

sns.lineplot(x=x, y=y, color="#2E86AB", linewidth=2)

plt.fill_between(x, y, color="#2E86AB", alpha=0.3)

plt.title("基础面积图")
plt.xlabel("序号")
plt.ylabel("数值")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
