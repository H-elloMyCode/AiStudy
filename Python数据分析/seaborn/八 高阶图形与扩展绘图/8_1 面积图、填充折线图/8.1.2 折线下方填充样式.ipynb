```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
x = np.linspace(1, 12, 30)
y = np.sin(x) + 6

x
```

```python
y
```

```python
plt.figure(figsize=(9, 4))

sns.lineplot(x=x, y=y, color="#0066cc", linewidth=2)

plt.title("折线下方自定义填充样式")
plt.xlabel("X轴")
plt.ylabel("Y轴")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4))

sns.lineplot(x=x, y=y, color="#0066cc", linewidth=2)

plt.fill_between(
    x,
    y,
    0,
    color="#0066cc",
    alpha=0.25,
    edgecolor="gray",
    linestyle="--"
)

plt.title("折线下方自定义填充样式")
plt.xlabel("X轴")
plt.ylabel("Y轴")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
