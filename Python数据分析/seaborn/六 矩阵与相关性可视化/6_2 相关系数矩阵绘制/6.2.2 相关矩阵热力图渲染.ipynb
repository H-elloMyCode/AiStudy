```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
corr_mat = tips.corr(numeric_only=True)

corr_mat
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr_mat,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("相关矩阵热力图渲染")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr_mat,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("相关矩阵热力图渲染")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr_mat,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.8
)

plt.title("相关矩阵热力图渲染")
plt.tight_layout()
plt.show()
```
