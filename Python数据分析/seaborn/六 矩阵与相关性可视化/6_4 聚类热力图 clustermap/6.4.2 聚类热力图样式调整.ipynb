```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
corr = tips.corr(numeric_only=True)

corr
```

```python
g = sns.clustermap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="RdBu_r",
    linewidths=0.8,
    figsize=(7, 6),
    col_cluster=True,  # 开启列聚类
    row_cluster=True,  # 开启行聚类
    cbar_kws={"shrink": 0.6}

)

# 调整刻度文字方向
plt.setp(g.ax_heatmap.get_xticklabels(), rotation=0)
plt.suptitle("聚类热力图样式调整", y=1.03)
plt.show()
```
