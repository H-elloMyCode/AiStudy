```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
corr = tips.corr(numeric_only=True)

corr
```

```python
# 聚类热力图
g = sns.clustermap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    figsize=(6, 5)
)

plt.setp(g.ax_heatmap.get_xticklabels(), rotation=0)
plt.suptitle("自动行列聚类热力图", y=1.02)
plt.show()
```
