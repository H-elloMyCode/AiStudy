```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
corr = tips.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("热力图配色条、刻度、标签优化")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r'
)

plt.title("热力图配色条、刻度、标签优化")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r',
    vmin=-1,
    vmax=1
)

plt.title("热力图配色条、刻度、标签优化")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r',
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title("热力图配色条、刻度、标签优化")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r',
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    xticklabels=True,
    yticklabels=True
)

plt.title("热力图配色条、刻度、标签优化")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r',
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    xticklabels=False,
    yticklabels=False
)

plt.title("热力图配色条、刻度、标签优化")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap='RdBu_r',
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    xticklabels=True,
    yticklabels=True,
    cbar_kws={"label": "相关系数", "shrink": 0.8}
)

plt.title("热力图配色条、刻度、标签优化")
plt.tight_layout()
plt.show()
```
