```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
corr = tips.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title("相关性矩阵热力图")
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(6, 5))

# 增加格子白线、固定色标范围-1~1，图表更规范
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, linewidths=0.6)

plt.title("相关性矩阵热力图")
plt.tight_layout()

plt.show()
```
