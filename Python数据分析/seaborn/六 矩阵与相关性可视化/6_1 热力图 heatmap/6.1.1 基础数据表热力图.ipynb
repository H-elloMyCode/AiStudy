```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
data = np.random.rand(5, 5)

data
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    data,
    annot=True,
    cmap="Blues"
)

plt.title("基础数据表热力图")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    data,
    # annot=True,
    cmap="Blues"
)

plt.title("基础数据表热力图")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    data,
    annot=True,
    cmap="Greens"
)

plt.title("基础数据表热力图")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    data,
    annot=True,
    cmap="Reds"
)

plt.title("基础数据表热力图")
plt.tight_layout()
plt.show()
```
