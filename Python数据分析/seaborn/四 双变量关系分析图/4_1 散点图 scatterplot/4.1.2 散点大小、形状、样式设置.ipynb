```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_style("whitegrid")

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("散点大小、形状、样式自定义")
plt.xlabel("总消费金额")
plt.ylabel("小费金额")

plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    s=80
)

plt.title("散点大小、形状、样式自定义")
plt.xlabel("总消费金额")
plt.ylabel("小费金额")

plt.show()
```
