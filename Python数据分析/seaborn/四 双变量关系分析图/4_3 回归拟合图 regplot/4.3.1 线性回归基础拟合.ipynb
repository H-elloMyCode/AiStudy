```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_style("whitegrid")

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(7, 4))

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    color="#2E86AB",
    line_kws={"color": 'red'}
)


plt.title("消费金额 vs 小费 线性回归拟合")
plt.show()
```
