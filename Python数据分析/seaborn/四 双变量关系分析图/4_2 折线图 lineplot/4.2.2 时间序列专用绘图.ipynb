```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
sns.set_style("whitegrid")

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
date_range = pd.date_range(start="2025-01-01", periods=60, freq="D")

date_range
```

```python
value = np.random.randn(60).cumsum() + 50

value
```

```python
df = pd.DataFrame({"date": date_range, "value": value})

df
```

```python
plt.figure(figsize=(9, 4))

sns.lineplot(data=df, x="date", y="value", color="#3498db", linewidth=2)

# 旋转日期避免重叠
plt.xticks(rotation=30)
plt.title("时间序列趋势图（日期 -> 数值）")
plt.tight_layout()

plt.show()
```
