```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.DataFrame(
    {
        "品类": ["苹果", "香蕉", "橙子", "葡萄"],
        "销量": [55, 40, 65, 70]
    }
)

df
```

```python
my_color_list = ["#FF6B6B", "#4ECDC4", "#45B7D1", "#FFA07A"]

sns.set_palette(my_color_list)
```

```python
sns.barplot(data=df, hue="品类", x="品类", y="销量")
plt.title("手动自定义调色板")
plt.show()
```

```python
my_color_list = ["#45B7D1", "#FFA07A", "#FF6B6B", "#4ECDC4"]

sns.set_palette(my_color_list)

sns.barplot(data=df, hue="品类", x="品类", y="销量")
plt.title("手动自定义调色板")
plt.show()
```

```python
my_color_list = ["red", "green", "yellow", "blue"]

sns.set_palette(my_color_list)

sns.barplot(data=df, hue="品类", x="品类", y="销量")
plt.title("手动自定义调色板")
plt.show()
```
