```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.DataFrame(
    {
        "姓名": ["张三", "李四", "王五", "赵六", "钱七"],
        "得分": [58, 72, 89, 65, 93]
    }
)

df
```

```python
plt.figure(figsize=(7, 4))

sns.barplot(
    data=df,
    x="姓名",
    hue="姓名",
    y="得分",
    palette="Blues"
)

plt.title("数值与颜色自动映射（数值越大→颜色越深）")
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.barplot(
    data=df,
    x="姓名",
    hue="姓名",
    y="得分",
    palette="Greens"
)

plt.title("数值与颜色自动映射（数值越大→颜色越深）")
plt.show()
```

```python
plt.figure(figsize=(7, 4))

# 1. 创建颜色映射
norm = plt.Normalize(df['得分'].min(), df['得分'].max())
cmap = plt.cm.Blues
colors = cmap(norm(df['得分']))

# 2. 用plt.bar画图，直接指定每个柱子的颜色
plt.bar(df['姓名'], df['得分'], color=colors)
plt.title("数值与颜色自动映射（数值越大→颜色越深）")
plt.ylabel("得分")
plt.show()
```

```python
# 先按得分从小到大排序
df_sorted = df.sort_values('得分').reset_index(drop=True)

plt.figure(figsize=(7, 4))
sns.barplot(
    data=df_sorted,
    x="姓名",
    hue="姓名",
    y="得分",
    palette="Blues"
)
plt.title("数值与颜色自动映射（数值越大→颜色越深）")
plt.show()
```
