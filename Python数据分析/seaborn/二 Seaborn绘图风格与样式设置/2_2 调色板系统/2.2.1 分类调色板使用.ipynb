```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sns.set_palette("colorblind")
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.DataFrame(
    {
        "班级": ["一班","二班","三班","四班","五班"],
        "平均分": [78,85,92,88,90]
    }
)

df
```

```python
sns.barplot(data=df, hue="班级", x="班级", y="平均分")
plt.title("分类调色板演示")
plt.show()
```
