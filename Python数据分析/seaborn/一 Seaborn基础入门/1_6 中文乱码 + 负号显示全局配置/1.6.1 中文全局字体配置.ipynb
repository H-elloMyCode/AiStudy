```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# plt.rcParams["font.sans-serif"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
sns.set_theme()
plt.rcParams["font.sans-serif"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
```

```python
df = pd.DataFrame({"类别": ["A", "B", "C"], "数值": [15, 22, 18]})
df
```

```python
sns.barplot(data=df, x="类别", y="数值")

plt.title("中文标题正常显示")
plt.xlabel("分类")
plt.ylabel("数量")

plt.show()
```
