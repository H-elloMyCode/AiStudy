```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
sns.set_theme()
plt.rcParams["font.sans-serif"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = np.linspace(-5, 5, 100)
y = -np.sin(x)
```

```python
sns.lineplot(x = x, y = y)
plt.title("负号正常显示测试")
plt.show()
```
