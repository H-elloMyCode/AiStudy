```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame(
    {
        "类别": ["A", "A", "B", "B", "C", "C"],
        "数值": [12, 15, 9, 11, 18, 20]
    }
)

df
```

```python
sns.set_theme()
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
plt.figure(figsize=(6, 4))

sns.barplot(data=df, x='类别', y='数值')
plt.title("Seaborn 一键分类绘图")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. 构造数据
df = pd.DataFrame({
    "类别": ["A", "A", "B", "B", "C", "C"],
    "数值": [12, 15, 9, 11, 18, 20]
})

# 2. 按类别分组计算均值和标准差
grouped = df.groupby("类别")["数值"]
x_labels = ["A", "B", "C"]

# ✅ 这里修复成 get_group()
means = [grouped.get_group(x).mean() for x in x_labels]
stds = [grouped.get_group(x).std() for x in x_labels]

# 3. 中文显示
plt.rcParams["font.family"] = ["SimHei", "Microsoft YaHei"]
plt.rcParams["axes.unicode_minus"] = False

# 4. 画图
plt.figure(figsize=(6, 4))
plt.bar(x=x_labels, height=means, yerr=stds, capsize=5, color="#1f77b4")

plt.title("Matplotlib 绘制分类柱状图")
plt.tight_layout()
plt.show()
```
