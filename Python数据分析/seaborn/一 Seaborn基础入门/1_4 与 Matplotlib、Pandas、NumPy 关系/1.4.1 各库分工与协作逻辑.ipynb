```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme()
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
x = np.linspace(0, 10, 50)
y = x * 2 + np.random.randn(50)
```

```python
df = pd.DataFrame({"X轴": x, "Y轴": y})
df
```

```python
sns.scatterplot(data=df, x="X轴", y="Y轴")

plt.title("四库协作演示")
plt.show()
```
