```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.DataFrame({"年龄": [20, 21, 22, 23, 24], "分数": [85, 78, 92, 88, 95]})
```

```python
df
```

```python
df = pd.DataFrame({"年龄": [20, 21, 22, 23, 24], "分数": [85, 78, 92, 88, 95]})

sns.lineplot(data=df, x="年龄", y="分数", label="DataFrame(最优)")

x_np = np.array([20, 21, 22, 23, 24])
y_np = np.array([70, 75, 80, 83, 87])
sns.lineplot(x=x_np, y=y_np, label="NumPy数组")

x_list = [20, 21, 22, 23, 24]
y_list = [60, 65, 70, 76, 82]
sns.lineplot(x=x_list, y=y_list, label="Python列表")

plt.title("三种数据格式对比")
plt.legend()
plt.show()
```
