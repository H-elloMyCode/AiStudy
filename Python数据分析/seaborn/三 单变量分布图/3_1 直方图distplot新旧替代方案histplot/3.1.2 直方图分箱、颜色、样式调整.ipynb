```python
import seaborn as sns
import matplotlib.pyplot as plt
from pyparsing import alphas
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(7, 4))

sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    color="#4ECDC4",
    edgecolor="black",
    alpha=0.7,
    kde=True
)

plt.show()
```
