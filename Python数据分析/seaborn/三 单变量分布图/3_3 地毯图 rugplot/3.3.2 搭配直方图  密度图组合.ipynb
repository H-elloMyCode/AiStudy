```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(8, 4))

sns.histplot(data=tips, x="total_bill", bins=15, alpha=0.3, color="skyblue", stat="density")

sns.kdeplot(data=tips, x="total_bill", linewidth=2, color="darkred")

sns.rugplot(data=tips, x="total_bill", color="black", height=0.1)

plt.title("消费金额：直方图 + 密度曲线 + 地毯图 组合")
plt.tight_layout()
plt.show()
```

```python
# 3. 创建画布
plt.figure(figsize=(10, 6))

# 4. 绘制直方图（主Y轴：Count）
ax1 = plt.gca()
sns.histplot(data=tips, x="total_bill", bins=15, alpha=0.3, color="skyblue", ax=ax1)
ax1.set_ylabel('Count')

# 5. 绘制密度曲线（次Y轴：Density）
ax2 = ax1.twinx()
sns.kdeplot(data=tips, x="total_bill", linewidth=2, color="darkred", ax=ax2)
ax2.set_ylabel('Density')

# 6. 绘制地毯图
sns.rugplot(data=tips, x="total_bill", color="black", height=0.1, ax=ax1)

# 7. 设置标题和布局
plt.title("消费金额：直方图 + 密度曲线 + 地毯图 组合")
plt.tight_layout()
plt.show()
```
