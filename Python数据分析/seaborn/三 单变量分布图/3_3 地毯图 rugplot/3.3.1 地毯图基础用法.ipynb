```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize=(7, 4))

sns.kdeplot(data=tips, x="total_bill", fill=True, alpha=0.4)

sns.rugplot(data=tips, x="total_bill", color="red", height=0.1)

plt.show()
```
