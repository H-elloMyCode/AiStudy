```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
```

```python
tips = sns.load_dataset("tips", data_home="D:\Projects\AiStudy\Python数据分析\seaborn\seaborn-data-master")

tips
```

```python
plt.figure(figsize = (7, 4))

sns.histplot(
    data=tips,
    x="total_bill",
    bins=15,
    color="skyblue",
    kde=True
)

plt.title("消费金额分布：直方图+密度曲线叠加")
plt.show()
```

```python
plt.figure(figsize = (7, 4))

sns.histplot(
    data=tips,
    x="total_bill",
    bins=15,
    color="skyblue"
)

plt.title("消费金额分布：直方图+密度曲线叠加")
plt.show()
```
