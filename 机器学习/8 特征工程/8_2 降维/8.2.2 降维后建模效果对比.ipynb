```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
import numpy as np
```

```python
X, y = load_iris(return_X_y=True)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
model = LogisticRegression(max_iter=200, random_state=42)

scores_4d = cross_val_score(model, X_scaled, y, cv=5, scoring='accuracy')
```

```python
scores_4d
```

```python
pca_2d = PCA(n_components=2)

X_pca_2d = pca_2d.fit_transform(X_scaled)
scores_2d = cross_val_score(model, X_pca_2d, y, cv=5, scoring='accuracy')
```

```python
scores_2d
```

```python
pca_3d = PCA(n_components=3)
X_pca_3d = pca_3d.fit_transform(X_scaled)
scores_3d = cross_val_score(model, X_pca_3d, y, cv=5, scoring='accuracy')
```

```python
scores_3d
```

```python
print("=" * 45)
print(f"{'方案':<20} {'维度':<6} {'准确率':<12} {'方差'}")
print("=" * 45)
print(f"{'原始数据':<20} {'4D':<6} {scores_4d.mean():.4f} ± {scores_4d.std():.4f}")
print(f"{'PCA 降到 3 维':<20} {'3D':<6} {scores_3d.mean():.4f} ± {scores_3d.std():.4f}")
print(f"{'PCA 降到 2 维':<20} {'2D':<6} {scores_2d.mean():.4f} ± {scores_2d.std():.4f}")
print("=" * 45)
```

```python
print(f"\nPCA(2D) 累计解释方差比: {pca_2d.explained_variance_ratio_.sum():.4f}")
print(f"PCA(3D) 累计解释方差比: {pca_3d.explained_variance_ratio_.sum():.4f}")
```
