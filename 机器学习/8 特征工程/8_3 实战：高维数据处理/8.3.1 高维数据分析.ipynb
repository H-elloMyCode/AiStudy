```python
from sklearn.datasets import make_classification
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X, y = make_classification(
    n_samples=500,
    n_features=20,
    n_informative=5,
    n_redundant=10,
    n_repeated=0,
    n_classes=2,
    random_state=42
)
```

```python
X
```

```python
print(f"数据形状: {X.shape}")
print(f"样本数: {X.shape[0]}, 特征数: {X.shape[1]}")
print(f"类别分布: {pd.Series(y).value_counts().to_dict()}\n")
```

```python
feature_names = [f'f{i}' for i in range(X.shape[1])]
```

```python
feature_names
```

```python
df = pd.DataFrame(X, columns=feature_names)
df['target'] = y
```

```python
df
```

```python
print("各特征基本统计:")
print(df.describe().loc[['mean', 'std', 'min', 'max']].T.head(10).round(2))
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, axes = plt.subplots(2, 3, figsize=(12, 6))

for i, ax in enumerate(axes.flat):
    for label in [0, 1]:
        mask = y == label
        ax.hist(X[mask, i], bins=30, alpha=0.5, label=f'类别{label}')

    ax.set_title(f'{feature_names[i]}')
    ax.legend(fontsize=8)

plt.suptitle('前 6 个特征的分布')
plt.tight_layout()
plt.savefig('high_dim_distribution.png', dpi=100)

plt.show()
```
