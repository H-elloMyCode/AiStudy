```python
from sklearn.datasets import make_classification
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.decomposition import PCA
import pandas as pd
```

```python
X, y = make_classification(
    n_samples=500, n_features=20,
    n_informative=5, n_redundant=10, random_state=42
)

feature_names = [f'f{i}' for i in range(X.shape[1])]
```

```python
X
```

```python
len(y)
```

```python
feature_names
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
print(f"原始维度: {X_scaled.shape[1]}\n")
```

```python
selector = SelectKBest(score_func=f_classif, k=10)
X_selected = selector.fit_transform(X_scaled, y)
```

```python
X_selected
```

```python
selected_features = [f for f, s in zip(feature_names, selector.get_support()) if s]
```

```python
selected_features
```

```python
print(f"SelectKBest 选择后维度: {X_selected.shape[1]}")
print(f"选中的特征: {selected_features}\n")
```

```python
scores = pd.Series(selector.scores_, index=feature_names).sort_values(ascending=False)
```

```python
scores
```

```python
print("特征得分 Top 10:")
print(scores.head(10).round(2).to_string())
print()
```

```python
pca = PCA(n_components=5)

X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca
```

```python
print(f"PCA 降维后维度: {X_pca.shape[1]}")
print(f"累计解释方差比: {pca.explained_variance_ratio_.sum():.4f}\n")
```

```python
X_selected_then_pca = PCA(n_components=5).fit_transform(X_selected)
X_selected_then_pca
```

```python
print(f"SelectKBest(10) + PCA(5) 最终维度: {X_selected_then_pca.shape[1]}")
```

```python
# ---- 汇总对比 ----
print("\n" + "=" * 45)
print(f"{'处理阶段':<25} {'维度':<6}")
print("=" * 45)
print(f"{'原始数据':<25} {X_scaled.shape[1]}")
print(f"{'SelectKBest':<25} {X_selected.shape[1]}")
print(f"{'PCA 直接降维':<25} {X_pca.shape[1]}")
print(f"{'SelectKBest + PCA':<25} {X_selected_then_pca.shape[1]}")
print("=" * 45)
```
