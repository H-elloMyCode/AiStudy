```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
```

```python
digits = load_digits()
X, y = digits.data, digits.target
```

```python
print(f"像素值范围: [{X.min()}, {X.max()}]")  # [0, 16]
print(f"像素值类型: {X.dtype}")
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

fig, axes = plt.subplots(2, 8, figsize=(12, 5))

for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap="gray_r")
    ax.set_title(f"Label: {y[i]}", fontsize=10)
    ax.axis("off")

plt.suptitle("手写数字样本 (8x8)", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 10, figsize=(15, 2))
for digit in range(10):
    idx = list(y).index(digit)  # 找到该数字的第一个样本
    axes[digit].imshow(digits.images[idx], cmap="gray_r")
    axes[digit].set_title(f"{digit}", fontsize=14)
    axes[digit].axis("off")

plt.suptitle("每个数字各一个样本", fontsize=14)
plt.tight_layout()
plt.show()
```
