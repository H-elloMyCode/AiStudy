```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
```

```python
k_range = range(1, 21)

train_scores = []
test_scores = []

for k in k_range:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)

    train_scores.append(accuracy_score(y_train, model.predict(X_train)))
    test_scores.append(accuracy_score(y_test, model.predict(X_test)))

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

plt.figure(figsize=(10, 5))

plt.plot(k_range, train_scores, 'o-', label='训练集准确率')
plt.plot(k_range, test_scores, 'o-', label='测试集准确率')
plt.xlabel('K值')
plt.ylabel('准确率')
plt.title('K值对KNN分类准确率的影响')
plt.xticks(k_range)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
best_k = list(k_range)[test_scores.index(max(test_scores))]
print(f"最佳K值: {best_k}，测试集准确率: {max(test_scores):.4f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 中文显示配置
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# 加载数据
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)

# 扩大K范围到1~100
k_range = range(1, 101)
train_scores = []
test_scores = []

for k in k_range:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)
    train_scores.append(accuracy_score(y_train, train_pred))
    test_scores.append(accuracy_score(y_test, test_pred))

# 画图展示K值对准确率的影响
plt.figure(figsize=(14, 6))
plt.plot(k_range, train_scores, 'o-', label='训练集准确率', markersize=4)
plt.plot(k_range, test_scores, 'o-', label='测试集准确率', markersize=4)
plt.xlabel('K值')
plt.ylabel('准确率')
plt.title('K值对KNN分类准确率的影响(K=1~100)')
plt.xticks(range(0, 101, 10))
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 找到最佳K值
best_k = list(k_range)[test_scores.index(max(test_scores))]
print(f"最佳K值: {best_k}，测试集准确率: {max(test_scores):.4f}")
```
