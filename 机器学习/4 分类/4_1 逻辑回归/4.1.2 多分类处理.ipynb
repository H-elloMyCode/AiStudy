```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)
```

```python
model_ovr = LogisticRegression(multi_class='ovr', max_iter=200)

model_ovr.fit(X_train, y_train)

print(f"OvR 准确率: {accuracy_score(y_test, model_ovr.predict(X_test)):.4f}")
```

```python
model_multi = LogisticRegression(multi_class='multinomial', max_iter=200)

model_multi.fit(X_train, y_train)

print(f"Multinomial  准确率: {accuracy_score(y_test, model_multi.predict(X_test)):.4f}")
```

```python
y_pred = model_multi.predict(X_test)
print("\n分类报告:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
```
