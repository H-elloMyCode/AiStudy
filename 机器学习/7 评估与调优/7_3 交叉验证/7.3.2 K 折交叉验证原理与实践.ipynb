```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import accuracy_score
import numpy as np
```

```python
X, y = load_iris(return_X_y=True)
```

```python
X
```

```python
len(y)
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
manual_scores = []
```

```python
for train_idx, test_idx in kf.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = RandomForestClassifier(random_state=42)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    manual_scores.append(accuracy_score(y_test, y_pred))

print("手动实现:")
print(f"  每次得分: {[f'{s:.4f}' for s in manual_scores]}")
print(f"  平均得分: {np.mean(manual_scores):.4f}")
```

```python
model = RandomForestClassifier(random_state=42)
auto_scores = cross_val_score(model, X, y, cv=kf, scoring="accuracy")

print("\ncross_val_score:")
print(f"  每次得分: {[f'{s:.4f}' for s in auto_scores]}")
print(f"  平均得分: {auto_scores.mean():.4f}")
```
