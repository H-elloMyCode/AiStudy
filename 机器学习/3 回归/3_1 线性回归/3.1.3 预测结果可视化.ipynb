```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([[50], [60], [70], [80], [90], [100], [110], [120]])
y = np.array([120, 150, 175, 200, 230, 260, 290, 320])
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
model
```

```python
print(f"斜率 coef_: {model.coef_}")
```

```python
print(f"截距 intercept_: {model.intercept_}")
```

```python
X_line = np.linspace(40, 130, 100).reshape(-1, 1)
y_line = model.predict(X_line)
```

```python
X_line
```

```python
y_line
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

plt.figure(figsize=(8, 5))

plt.scatter(X, y, color="steelblue", s=80, label='真实数据', zorder=3)

plt.plot(X_line, y_line, color='tomato', linewidth=2, label='回归线')

plt.xlabel('面积 (㎡)', fontsize=12)
plt.ylabel('房价 (万元)', fontsize=12)
plt.title('线性回归：面积 vs 房价', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
