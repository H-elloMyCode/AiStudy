```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
```

```python
np.random.seed(42)

X = np.sort(np.random.rand(30, 1) * 6, axis=0)
X
```

```python
y = np.sin(X).ravel() + np.random.randn(30) * 0.2
y
```

```python
len(y)
```

```python
plt.rcParams["font.family"] = ["DejaVu Sans", "SimHei", "Microsoft YaHei"]
plt.rcParams['axes.unicode_minus'] = False

degrees = [1, 3, 10]

plt.figure(figsize=(12, 4))

for i, degree in enumerate(degrees):
    ax = plt.subplot(1, 3, i + 1)

    model = make_pipeline(
        PolynomialFeatures(degree=degree, include_bias=False),
        LinearRegression()
    )

    model.fit(X, y)

    X_plot = np.linspace(0, 6, 200).reshape(-1, 1)
    y_plot = model.predict(X_plot)

    plt.scatter(X, y, color='steelblue', s=30, label='真实数据')
    plt.plot(X_plot, y_plot, color='tomato', linewidth=2, label=f'degree={degree}')

    plt.title(f'多项式 degree={degree}')
    plt.legend()
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
