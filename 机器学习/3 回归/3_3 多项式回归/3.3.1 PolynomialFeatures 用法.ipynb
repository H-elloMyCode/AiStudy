```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
```

```python
X = np.array([[2], [3]])
X
```

```python
poly2 = PolynomialFeatures(degree=2, include_bias=False)
X_poly2 = poly2.fit_transform(X)
```

```python
print(f"\ndegree=2 特征：")
print(X_poly2)
```

```python
poly3 = PolynomialFeatures(degree=3, include_bias=False)
X_poly3 = poly3.fit_transform(X)

print(f"\ndegree=3 特征：")
print(X_poly3)
```

```python
X_multi = np.array([[1, 2], [3, 4]])
X_multi
```

```python
poly2m = PolynomialFeatures(degree=2, include_bias=False)

X_poly_multi = poly2m.fit_transform(X_multi)

print(f"\n多特征 degree=2：")
print(f"特征名: {poly2m.get_feature_names_out()}")
print(X_poly_multi)
```

```python
X_multi = np.array([[1, 2], [3, 4]])

poly2m = PolynomialFeatures(degree=2, include_bias=False, interaction_only=True)

X_poly_multi = poly2m.fit_transform(X_multi)

print(f"\n多特征 degree=2：")
print(f"特征名: {poly2m.get_feature_names_out()}")
print(X_poly_multi)
```

```python
X_multi = np.array([[1, 2, 3], [4, 5, 6]])

poly2m = PolynomialFeatures(degree=2, include_bias=False)

X_poly_multi = poly2m.fit_transform(X_multi)

print(f"\n多特征 degree=2：")
print(f"特征名: {poly2m.get_feature_names_out()}")
print(X_poly_multi)
```
