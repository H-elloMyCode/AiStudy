```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import cross_val_score
from xgboost import XGBClassifier
import warnings

warnings.filterwarnings("ignore")
```

```python
X, y = load_breast_cancer(return_X_y=True)
```

```python
X
```

```python
len(y)
```

```python
print("=== max_depth 对比 ===")

for depth in [2, 4, 6, 8]:
    model = XGBClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=depth,
        eval_metric="logloss",
        random_state=42
    )

    scores = cross_val_score(model, X, y, cv=5, scoring='accuracy')
    print(f"  max_depth={depth}  准确率: {scores.mean():.4f} (±{scores.std():.4f})")
```

```python
print("\n=== subsample 对比 ===")

for sub in [0.6, 0.8, 1.0]:
    model = XGBClassifier(
        n_estimators=100, learning_rate=0.1, max_depth=3,
        subsample=sub, eval_metric='logloss', random_state=42
    )
    scores = cross_val_score(model, X, y, cv=5, scoring='accuracy')
    print(f"  subsample={sub}  准确率: {scores.mean():.4f} (±{scores.std():.4f})")
```

```python
print("\n=== learning_rate 对比 ===")

for lr in [0.01, 0.05, 0.1, 0.3]:
    model = XGBClassifier(
        n_estimators=100, learning_rate=lr, max_depth=3,
        eval_metric='logloss', random_state=42
    )
    scores = cross_val_score(model, X, y, cv=5, scoring='accuracy')
    print(f"  learning_rate={lr}  准确率: {scores.mean():.4f} (±{scores.std():.4f})")
```
