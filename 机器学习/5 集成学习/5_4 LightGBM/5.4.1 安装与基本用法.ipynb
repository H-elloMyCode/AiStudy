```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from lightgbm import LGBMClassifier
```

```python
X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
len(y_train)
```

```python
lgbm = LGBMClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=-1,
    num_leaves=31,
    verbose=-1,
    random_state=42
)
```

```python
lgbm.fit(X_train, y_train)
```

```python
y_pred = lgbm.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
print(f"LightGBM 准确率: {accuracy_score(y_test, y_pred):.4f}")
print()
print(classification_report(y_test, y_pred, target_names=['恶性', '良性']))
```

```python
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)
```

```python
X_val
```

```python
import lightgbm as lgb
from lightgbm import LGBMClassifier

lgbm = LGBMClassifier(
    n_estimators=500,
    learning_rate=0.1,
    num_leaves=31,
    verbose=-1,
    random_state=42
)

# 正确写法：从 lgb 取回调
callbacks = [
    lgb.early_stopping(stopping_rounds=50),
    lgb.log_evaluation(0)
]

lgbm.fit(
    X_tr, y_tr,
    eval_set=[(X_val, y_val)],
    callbacks=callbacks
)

print(f"最佳轮数: {lgbm.best_iteration_}")
```

```python
import numpy as np
import pandas as pd
from lightgbm import LGBMClassifier

np.random.seed(42)

df = pd.DataFrame({
    'city': np.random.choice(['北京', '上海', '广州', '深圳'], 500),
    'age': np.random.randint(18, 65, 500),
    'income': np.random.normal(15000, 5000, 500),
    'bought': np.random.choice([0, 1], 500)
})

# 关键：转为pandas分类类型
df['city'] = df['city'].astype('category')

X = df[['city', 'age', 'income']]
y = df['bought']

lgbm = LGBMClassifier(n_estimators=50, verbose=-1, random_state=42)
# 此时再指定分类列即可正常运行
lgbm.fit(X, y, categorical_feature=['city'])
```
