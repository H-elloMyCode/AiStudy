```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import cross_val_score
from lightgbm import LGBMClassifier
import warnings

warnings.filterwarnings('ignore')
```

```python
X, y = load_breast_cancer(return_X_y=True)
```

```python
# 不同 num_leaves 对比
print("=== num_leaves 对比 ===")

for nl in [15, 31, 63, 127]:
    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        num_leaves=nl,
        verbose=-1,
        random_state=42
    )

    scores = cross_val_score(model, X, y, cv=5, scoring='accuracy')
    print(f"  num_leaves={nl:>3d}  准确率: {scores.mean():.4f} (±{scores.std():.4f})")
```

```python
# 不同 max_depth 对比
print("\n=== max_depth 对比 ===")

for md in [3, 5, 7, -1]:
    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        num_leaves=31,
        max_depth=md,
        verbose=-1,
        random_state=42
    )

    scores = cross_val_score(model, X, y, cv=5, scoring='accuracy')
    label = md if md > 0 else "不限"
    print(f"  max_depth={label!s:>3s}  准确率: {scores.mean():.4f} (±{scores.std():.4f})")
```

```python
# 不同 subsample 对比
print("\n=== subsample 对比 ===")

for sub in [0.6, 0.8, 1.0]:
    model = LGBMClassifier(
        n_estimators=100,
        learning_rate=0.1,
        num_leaves=31,
        subsample=sub,
        verbose=-1,
        random_state=42
    )

    scores = cross_val_score(model, X, y, cv=5, scoring='accuracy')
    print(f"  subsample={sub}  准确率: {scores.mean():.4f} (±{scores.std():.4f})")
```
