```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
import numpy as np

# 加载数据
X, y = load_iris(return_X_y=True)

# 单棵决策树
tree = DecisionTreeClassifier(random_state=42)
tree_scores = cross_val_score(tree, X, y, cv=5, scoring='accuracy')

# Bagging：降低每棵树的训练样本、制造更多差异
bagging = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=30,        # 增加树数量
    max_samples=0.7,        # 每棵只抽70%样本
    max_features=0.7,       # 每棵只抽70%特征
    random_state=42
)
bagging_scores = cross_val_score(bagging, X, y, cv=5, scoring='accuracy')

# 对比结果
print(f"单棵决策树  平均准确率: {tree_scores.mean():.4f}  (±{tree_scores.std():.4f})")
print(f"Bagging(30) 平均准确率: {bagging_scores.mean():.4f}  (±{bagging_scores.std():.4f})")
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
import numpy as np

# 加载数据
X, y = load_iris(return_X_y=True)

# 单棵决策树
tree = DecisionTreeClassifier(random_state=42)
tree_scores = cross_val_score(tree, X, y, cv=5, scoring='accuracy')

# Bagging：降低每棵树的训练样本、制造更多差异
bagging = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=30,
    max_samples=0.85,   # 少删一点样本
    max_features=1.0,   # 全部特征不截断
    random_state=42
)
bagging_scores = cross_val_score(bagging, X, y, cv=5, scoring='accuracy')

# 对比结果
print(f"单棵决策树  平均准确率: {tree_scores.mean():.4f}  (±{tree_scores.std():.4f})")
print(f"Bagging(30) 平均准确率: {bagging_scores.mean():.4f}  (±{bagging_scores.std():.4f})")
```

```python
for n in [1, 5, 10, 50, 100]:
    bagging = BaggingClassifier(
        estimator=DecisionTreeClassifier(),
        n_estimators=n,
        random_state=42
    )
    scores = cross_val_score(bagging, X, y, cv=5, scoring='accuracy')
    print(f"n_estimators={n:>3d}  准确率: {scores.mean():.4f}  (±{scores.std():.4f})")
```
