```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
X_train
```

```python
y_train
```

```python
len(y_train)
```

```python
gbdt = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    subsample=0.8,
    random_state=42
)
```

```python
gbdt.fit(X_train, y_train)
```

```python
y_pred = gbdt.predict(X_test)
```

```python
y_test
```

```python
y_pred
```

```python
print(f"GBDT 准确率: {accuracy_score(y_test, y_pred):.4f}")
print()
print(classification_report(y_test, y_pred, target_names=['恶性', '良性']))
```

```python
import matplotlib.pyplot as plt
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
train_losses = []
test_losses = []

for y_train_pred, y_test_pred in zip(gbdt.staged_predict(X_train), gbdt.staged_predict(X_test)):
    train_losses.append(1 - accuracy_score(y_train, y_train_pred))
    test_losses.append(1 - accuracy_score(y_test, y_test_pred))

plt.figure(figsize=(8, 4))
plt.plot(train_losses, label='训练误差')
plt.plot(test_losses, label='测试误差')

plt.xlabel('树的数量')
plt.ylabel('误差')
plt.title('GBDT 训练过程')
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
