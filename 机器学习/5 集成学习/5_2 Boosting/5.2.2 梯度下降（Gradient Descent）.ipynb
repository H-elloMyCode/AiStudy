```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

X = np.random.rand(100, 1) * 10
X
```

```python
y = 2 * X + 1 + np.random.randn(100, 1) * 2
```

```python
y
```

```python
w = np.random.randn(1, 1)
w
```

```python
b = np.random.randn(1, 1)
b
```

```python
lr = 0.01

losses = []
```

```python
for epoch in range(100):
    y_pred = X * w + b

    loss = np.mean((y_pred - y) ** 2)
    losses.append(loss)

    dw = np.mean(2 * (y_pred - y) * X)
    db = np.mean(2 * (y_pred - y))

    w -= lr * dw
    b -= lr * db

print(f"学到的参数: w = {w[0][0]:.2f}, b = {b[0][0]:.2f}")
print(f"真实参数:   w = 2.00, b = 1.00")
```

```python
plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(losses)
plt.xlabel('迭代轮数')
plt.ylabel('MSE 损失')
plt.title('梯度下降：损失逐渐下降')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.scatter(X, y, alpha=0.5, label='真实值')
x_line = np.linspace(0, 10, 100).reshape(-1, 1)
plt.plot(x_line, x_line * w + b, color='red', label='梯度下降拟合')
plt.xlabel('X')
plt.ylabel('y')
plt.title('拟合效果')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
