```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
n = 200
```

```python
data = {
    '年龄': np.concatenate([
        np.random.randint(18, 35, 70),
        np.random.randint(30, 50, 80),
        np.random.randint(40, 65, 50),
    ]),
    '年收入_万': np.concatenate([
        np.random.uniform(10, 30, 70),
        np.random.uniform(40, 80, 80),
        np.random.uniform(20, 45, 50),
    ]),
    '消费分数': np.concatenate([
        np.random.uniform(60, 100, 70),
        np.random.uniform(60, 95, 80),
        np.random.uniform(10, 40, 50),
    ]),
}
```

```python
df = pd.DataFrame(data)
```
