```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.6, random_state=42)

```

```python
X
```

```python
y_true
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

```

```python
kmeans.fit(X)
```

```python
print("簇标签:", kmeans.labels_[:10])
```

```python
kmeans.labels_
```

```python
len(kmeans.labels_)
```

```python
print("质心坐标:\n", kmeans.cluster_centers_)
```

```python
print("惯性值:", kmeans.inertia_)
```
