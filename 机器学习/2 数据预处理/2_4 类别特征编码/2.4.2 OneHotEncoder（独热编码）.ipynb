```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    '城市': ['北京', '上海', '深圳', '北京', '上海'],
    '年龄': [25, 30, 35, 28, 42],
})
```

```python
df
```

```python
encoder = OneHotEncoder(sparse_output=False)

city_encoded = encoder.fit_transform(df[['城市']])
city_encoded
```

```python
print(encoder)
```

```python
print("编码类别:", encoder.categories_)
print("\n独热编码结果:")
print(pd.DataFrame(city_encoded, columns=encoder.get_feature_names_out(['城市'])))
```
