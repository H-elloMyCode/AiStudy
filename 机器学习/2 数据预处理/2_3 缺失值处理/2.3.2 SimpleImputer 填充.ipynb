```python
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
```

```python
df = pd.DataFrame({
    '年龄': [25, np.nan, 35, 28, np.nan, 42, 31, 29],
    '收入': [8000, 15000, np.nan, 12000, 9500, np.nan, 11000, 13000],
})

df
```

```python
print(df)
```

```python
imputer_mean = SimpleImputer(strategy='mean')

df_mean = pd.DataFrame(imputer_mean.fit_transform(df), columns=df.columns)

df_mean
```

```python
df_mean.round(1)
```

```python
imputer_median = SimpleImputer(strategy='median')

df_median = pd.DataFrame(imputer_median.fit_transform(df), columns=df.columns)

df_median
```

```python
df_median.round(1)
```

```python
imputer_most_frequent = SimpleImputer(strategy='most_frequent')

df_most_frequent = pd.DataFrame(imputer_most_frequent.fit_transform(df), columns=df.columns)

df_most_frequent
```

```python
imputer_constant = SimpleImputer(strategy='constant', fill_value=10000)

df_constant = pd.DataFrame(imputer_constant.fit_transform(df), columns=df.columns)

df_constant
```

```python
import pandas as pd

df_city = pd.DataFrame({'城市': ['北京', '上海', None, '深圳', '北京', None]})
# 取出众数值填充
mode_val = df_city['城市'].mode()[0]
df_city['城市'] = df_city['城市'].fillna(mode_val)
print(df_city)
```

```python
df_city
```

```python
# 用众数填充类别特征
df_city = pd.DataFrame({'城市': ['北京', '上海', np.nan, '深圳', '北京', np.nan]})
```

```python
df_city
```

```python
imputer_mode = SimpleImputer(strategy='most_frequent')
# 加 .ravel() / .flatten() 把二维转一维
df_city['城市'] = imputer_mode.fit_transform(df_city[['城市']]).ravel()

df_city
```
