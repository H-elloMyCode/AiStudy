```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
data = np.array([
    [25, 8000],
    [30, 15000],
    [35, 20000],
    [45, 35000],
    [55, 50000],
])
```

```python
scaler = MinMaxScaler()

data_scaled = scaler.fit_transform(data)

data_scaled
```

```python
print("原始数据:")
print(data)
print("\n归一化后（缩放到 [0,1]）:")
print(np.round(data_scaled, 2))
print(f"\n最小值: {scaler.data_min_}")
print(f"最大值: {scaler.data_max_}")
```
