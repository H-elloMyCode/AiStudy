```python
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
data = np.array([
    [25, 8000],
    [30, 15000],
    [35, 20000],
    [45, 35000],
    [55, 50000],
])

data
```

```python
scaler = StandardScaler()

data_scaled = scaler.fit_transform(data)

data_scaled
```

```python
print("原始数据:")
print(data)
print("\n标准化后:")
print(np.round(data_scaled, 2))
print(f"\n均值: {scaler.mean_}")
print(f"标准差: {np.sqrt(scaler.var_)}")
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test = train_test_split(data, test_size=0.4, random_state=42)
```
